##### ✍️ 방향에 맞춰 이동

https://www.codetree.ai/ko/external-connection/classes/103/lectures/930/curated-cards/intro-move-in-direction/description


```python
n = int(input())
moves = [tuple(input().split()) for _ in range(n)]
dir = [move[0] for move in moves]
dist = [int(move[1]) for move in moves]

x, y = 0, 0


dx = [0, 0, -1, 1]
dy = [-1, 1, 0, 0]

for i in range(n):
    if dir[i] == 'N':
        x += dx[1] * dist[i]
        y += dy[1] * dist[i]

    elif dir[i] == 'S':
        x += dx[0] * dist[i]
        y += dy[0] * dist[i]
    
    elif dir[i] == 'E':
        x += dx[3] * dist[i]
        y += dy[3] * dist[i]

    else:
        x += dx[2] * dist[i]
        y += dy[2] * dist[i]

print(x,y)
```

##### ✍️ 문자에 따른 명령 2

https://www.codetree.ai/ko/external-connection/classes/103/lectures/930/curated-cards/intro-text-based-commands2/description

```python
dirs = input()

x, y = 0, 0

dx = [-1, 0, 1, 0]
dy = [0, 1, 0, -1]
direction = 1
for i in range(len(dirs)):
    if dirs[i] == 'L':
        direction -= 1
        direction %= 4
    elif dirs[i] =='F':
        x += dx[direction]
        y += dy[direction]
    else:
        direction += 1
        direction %= 4
        
print(x, y)
    

```
