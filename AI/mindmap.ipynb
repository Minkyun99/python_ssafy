# 🧠 머신러닝 (Machine Learning) 핵심 마인드맵

머신러닝은 데이터를 통해 학습하고 예측이나 의사결정을 수행하는 알고리즘을 구축하는 학문입니다. 크게 **지도 학습(Supervised Learning)**과 **비지도 학습(Unsupervised Learning)**으로 나뉩니다.

I. 지도 학습 (Supervised Learning)

정의: 입력 데이터($X$)와 그에 해당하는 정답 레이블($Y$) 쌍을 학습하여, 새로운 데이터에 대한 $Y$를 예측하는 모델을 구축합니다.

1. 주요 목표 및 개념

목표 | 세부 목표 | 설명
|--|--|--|
회귀 (Regression) | 연속적인 값 예측 | 주택 가격, 온도, 판매량 등 연속적인 수치 예측
분류 (Classification) | 범주형 값 예측 | 이메일 스팸 여부(이진), 꽃의 종류(다중 범주) 등 이산적인 레이블 예측

2. 주요 알고리즘, 손실 함수, 및 활성화 함수 (추가/정리됨)

목표 | 주요 알고리즘 | 사용 손실 함수 | 최종 출력층 활성화 함수
|--|--|--|--|
회귀 | 선형 회귀 (Linear Regression) | MSE (Mean Squared Error) | None (선형 출력)
회귀 | 릿지/라쏘 회귀 (Ridge/Lasso) | MSE + L1/L2 규제 | None (선형 출력)
분류 | 로지스틱 회귀 (Logistic Regression) | BCE (Binary Cross-Entropy) | 시그모이드 (Sigmoid)
분류 | 서포트 벡터 머신 (SVM) | 힌지 손실 (Hinge Loss) | None (부호에 따라 분류)
분류 | 결정 트리/랜덤 포레스트 | 엔트로피 (정보 이득) 또는 지니 불순도 (Gini) | None (트리 기반은 직접 확률/클래스 출력)
분류 | 인공신경망 (ANN/DNN) | BCE (이진), CCE (다중) | 시그모이드 (이진), 소프트맥스 (Softmax) (다중)

3. 주요 손실 함수 (Loss Function) 및 목적

모델의 예측($\hat{y}$)이 실제 정답($y$)과 얼마나 차이가 나는지 측정합니다.

종류 | 수식 | 목적
|--|--|--|
평균 제곱 오차 (MSE) | $$\frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2$$ | 회귀 모델의 성능 지표 (오차의 제곱 평균)
이진 교차 엔트로피 (BCE) | $$-[y \log(\hat{y}) + (1-y) \log(1-\hat{y})]$$ | 이진 분류 (예: 로지스틱 회귀) 모델의 성능 지표.
범주형 교차 엔트로피 (CCE) | $$-\sum_{i=1}^{C}y_i \log(\hat{y}_i)$$ | 다중 분류 (예: 소프트맥스 출력) 모델의 성능 지표

4. 주요 활성화 함수 (Activation Function)

신경망의 비선형성을 부여하고, 출력을 특정 범위로 변환합니다.

종류 | 수식 | 목적
|--|--|--|
시그모이드 (Sigmoid) | $$\sigma(z) = \frac{1}{1 + e^{-z}}$$ | 출력을 $0$과 $1$ 사이의 확률로 변환. 주로 이진 분류의 최종 출력층에 사용.
ReLU | $$\text{ReLU}(z) = \max(0, z)$$ | 숨겨진 층(Hidden Layer)에서 가장 흔하게 사용. 계산이 빠르고 경사 소실 문제를 완화.
소프트맥스 (Softmax) | $$\text{Softmax}(z)_i = \frac{e^{z_i}}{\sum_{j=1}^{C}e^{z_j}}$$ | 다중 분류의 최종 출력층에 사용. 모든 클래스의 확률 합을 $1$로 만듭니다.

II. 비지도 학습 (Unsupervised Learning)

정의: 정답 레이블($Y$)이 없는 데이터($X$)만을 학습하여, 데이터 내부에 존재하는 숨겨진 패턴, 구조, 특징을 발견합니다.

1. 주요 목표 및 개념

목표 | 세부 목표 | 설명
|--|--|--|
군집화 (Clustering) | 유사한 데이터끼리 그룹화 | 고객 세분화, 이미지 분할 등
차원 축소 (Dimensionality Reduction) | 데이터의 특징(Feature) 수 감소 | 데이터 시각화, 계산 효율성 증대
연관 규칙 (Association) | 항목 간의 숨겨진 관계 발견 | 장바구니 분석(함께 구매하는 품목 찾기)

2. 주요 알고리즘 및 목적 함수 (활성화 함수 정보 추가됨)

목표 | 주요 알고리즘 | 주요 목적 함수/지표 | 활성화 함수 (Activation Function) | 설명
|--|--|--|--|--|
군집화 | K-평균 (K-Means) |  관성 (Inertia) (군집 중심과의 거리 최소화) | None (비신경망 모델) | 미리 정한 $K$개의 중심을 기준으로 데이터를 그룹화
군집화 | DBSCAN | $-$ (밀도 기반, 목적 함수 없음) | None (비신경망 모델) | 밀도 기반 군집화. 데이터 밀도가 높은 영역을 군집으로 식별
차원 축소 | 주성분 분석 (PCA) | 설명된 분산 최대화 | None (비신경망 모델) | 데이터의 분산이 가장 큰 방향(주성분)을 찾아 고차원 데이터를 저차원으로 투영
군집화 | 오토인코더 (Autoencoder)| 재구성 오류 최소화 ($\text{MSE}$ 등) | ReLU (은닉층), Sigmoid (출력층, 0-1 데이터 복원 시) | 신경망을 사용하여 입력 데이터를 압축 및 복원하여 잠재 특징을 학습
연관 규칙 | Apriori Algorithm | 지지도 (Support), 신뢰도 (Confidence) |None (비신경망 모델) | 데이터 집합에서 빈번하게 발생하는 항목 집합을 찾아내는 알고리즘

재구성 오류 (Reconstruction Error) | $$\frac{1}{n}\sum_{i=1}^{n} | -

# ⭐ 활성화 함수(Activation Function)를 사용하는 이유

활성화 함수는 신경망(Neural Network)의 각 뉴런(노드) 출력값을 결정하는 함수로, 모델이 복잡한 데이터 패턴을 학습할 수 있도록 하는 데 결정적인 역할을 합니다.

1. 비선형성 부여 (가장 중요한 이유)
    - 선형 모델의 한계: 만약 활성화 함수를 사용하지 않거나, 선형 함수($y = ax+b$)만 사용한다면, 아무리 층(layer)을 깊게 쌓더라도 이는 결국 하나의 큰 선형 함수로 표현됩니다.
    - 복잡한 문제 해결 불가능: 선형 함수는 XOR 문제나 이미지 분류와 같이 입력과 출력이 직선으로 구분될 수 없는 복잡한 비선형 관계를 절대로 학습할 수 없습니다.
    - 활성화 함수의 역할: ReLU, Sigmoid, Softmax와 같은 비선형 활성화 함수를 각 층에 적용함으로써, 신경망은 직선이 아닌 곡선, 즉 복잡한 비선형 의사결정 경계를 만들 수 있게 됩니다. 이는 신경망이 실제 세계의 복잡한 패턴(이미지, 음성, 자연어)을 모델링하고 학습할 수 있도록 합니다.

2. 기울기 제어 및 소실 방지
    - 경사 하강법: 신경망은 오차를 줄이기 위해 역전파(Backpropagation) 과정에서 미분값(기울기)을 계산하여 가중치를 업데이트합니다.
    - ReLU의 효율성: 활성화 함수는 이 기울기(Gradient)의 흐름을 제어합니다. 예를 들어, ReLU 함수($\text{ReLU}(z) = \max(0, z)$)는 양수 범위에서 기울기를 $1$로 유지하여, Sigmoid와 같은 함수에서 발생하던 기울기 소실(Vanishing Gradient) 문제를 완화하고 학습 속도를 개선합니다.

3. 출력값의 형태 조정데이터 정규화: 활성화 함수는 뉴런의 출력값을 특정 범위로 제한하거나 변환합니다.
    - Sigmoid: 출력값을 $0$과 $1$ 사이의 확률로 변환하여 이진 분류의 최종 출력층에 사용됩니다.
    - Softmax: 출력값을 여러 범주에 대한 확률 분포로 변환하여 다중 분류의 최종 출력층에 사용됩니다.
    - Hidden Layer: 은닉층의 활성화 함수(주로 ReLU)는 출력값의 범위를 제어하여 다음 층의 입력으로 전달될 때 값의 폭발적인 증가(Exploding Gradient)를 방지하는 데 도움을 줍니다.
    
-----------

요약하자면, 활성화 함수는 신경망이 단순한 선형 모델의 한계를 벗어나 세상의 복잡한 비선형 문제를 성공적으로 학습할 수 있도록 만드는 핵심적인 구성 요소입니다.참고로, machine_learning_mindmap.md 파일의 '주요 활성화 함수' 섹션에서 ReLU, Sigmoid, Softmax의 수식과 목적을 확인할 수 있습니다.

종류 | 수식| 목적
|--|--|--|
시그모이드 (Sigmoid) | $$\sigma(z) = \frac{1}{1 + e^{-z}}$$ | 출력을 $0$과 $1$ 사이의 확률로 변환. 주로 이진 분류의 최종 출력층에 사용.
ReLU | $$\text{ReLU}(z) = \max(0, z)$$ | 숨겨진 층(Hidden Layer)에서 가장 흔하게 사용. 계산이 빠르고 경사 소실 문제를 완화.
소프트맥스 (Softmax) | $$\text{Softmax}(z)_i = \frac{e^{z_i}}{\sum_{j=1}^{C}e^{z_j}}$$ | 다중 분류의 최종 출력층에 사용. 모든 클래스의 확률 합을 $1$로 만듭니다.

# 🧠 인공신경망 (Artificial Neural Network, ANN) 심화 학습

인공신경망은 인간의 뇌 구조를 모방하여 만든 계산 모델이며, 딥러닝(Deep Learning)의 기본 토대입니다.

1. 기본 구성 요소 및 구조

1.1. 뉴런 (Neuron) 또는 퍼셉트론 (Perceptron)

신경망의 가장 기본적인 처리 단위입니다. 여러 입력($\text{x}$)을 받아 가중치($\text{W}$)와 결합하여 하나의 출력($\text{a}$)을 만들어냅니다.

용어 | 기호 | 정의 및 역할
|--|--|--|
입력 (Input) | $x_1, x_2, \dots, x_n$ | 이전 뉴런의 출력값이거나 모델의 실제 입력 피처
가중치 (Weight) | $W_1, W_2, \dots, W_n$ | 입력 신호의 중요도를 나타내는 값. 학습 과정에서 조정됨.
편향 (Bias) | $b$ | 뉴런이 활성화되는 임계점을 조절하는 상수. 입력값과 무관하게 뉴런의 활성도를 조절함.
순입력 (Net Input) | $z$ | 입력($x$)과 가중치($W$)의 가중합에 편향($b$)을 더한 값.
활성화 출력 (Activation) | $a$ | 순입력을 활성화 함수 $f(\cdot)$에 통과시켜 얻는 최종 출력값.

$$\text{Net Input (z)} = \sum_{i} W_i x_i + b = W^T x + b$$

$$\text{Activation (a)} = f(z)$$

1.2. 계층 (Layer)

뉴런들이 모여 형성하는 구조입니다.

계층 | 역할
|--|--|
입력층 (Input Layer) | 외부 데이터를 받아들이는 층. 계산은 수행하지 않음.
은닉층 (Hidden Layer) | 입력 데이터에서 **복잡하고 비선형적인 특징(Feature)**을 추출하고 변환하는 층. 딥러닝은 은닉층이 2개 이상인 신경망을 의미합니다.
출력층 (Output Layer) | 모델의 최종 예측 결과를 내보내는 층. 문제 유형(회귀, 분류)에 따라 활성화 함수가 결정됨.

2. 핵심 활성화 함수 (Activation Function)

활성화 함수는 신경망에 비선형성을 부여하여 모델이 복잡한 패턴을 학습할 수 있도록 합니다.

종류 | 수식 | 목적 및 특징
|--|--|--|
시그모이드 (Sigmoid) | $$\sigma(z) = \frac{1}{1 + e^{-z}}$$ | 출력을 $\mathbf{0}$과 $\mathbf{1}$ 사이의 확률로 변환. 이진 분류의 최종 출력층에 사용. (단점: 기울기 소실 문제)
ReLU (Rectified Linear Unit) | $$\text{ReLU}(z) = \max(0, z)$$ | $0$보다 크면 값을 그대로 출력. 계산이 빠르고 기울기 소실 문제 완화. 은닉층에서 가장 널리 사용됨.
소프트맥스 (Softmax) | $$\text{Softmax}(z)_i = \frac{e^{z_i}}{\sum_{j=1}^{C}e^{z_j}}$$ | 다중 분류의 최종 출력층에 사용. 모든 클래스에 대한 확률의 합을 $\mathbf{1}$로 만듦.

3. 신경망의 학습 과정

신경망의 학습은 순전파로 오차를 계산하고 역전파로 가중치를 조정하는 반복적인 과정입니다.

3.1. 순전파 (Forward Propagation)

    1. 입력층의 데이터($x$)가 신경망의 각 층을 차례로 통과합니다.  

    2. 각 뉴런은 순입력 $z = W^T x + b$를 계산한 후, 활성화 함수 $f(\cdot)$를 통과하여 출력 $a$를 다음 층으로 전달합니다.  

    3. 마지막 출력층에서 최종 예측값 $\hat{y}$를 생성합니다.  

3.2. 손실 함수 (Loss Function)

예측값 $\hat{y}$와 실제 정답 $y$ 사이의 오차(Error)를 측정합니다. 이 값을 최소화하는 것이 학습의 목표입니다.

종류 | 수식 | 용도
|--|--|--|
평균 제곱 오차 (MSE) | $$\frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2$$ | 회귀 문제에서 주로 사용. 오차의 제곱 합을 최소화.
교차 엔트로피 (Cross-Entropy) | $$-[y \log(\hat{y}) + (1-y) \log(1-\hat{y})]$$ |  (이진) 분류 문제에서 사용. 예측 확률과 정답 확률 분포의 차이를 최소화.

3.3. 역전파 (Backpropagation)와 경사 하강법 (Gradient Descent)

    1. 역전파: 출력층에서 계산된 오차(손실 함수 값)를 입력층 방향(뒤에서 앞으로)으로 역행하면서 각 가중치($W$)와 편향($b$)이 전체 오차에 얼마나 기여했는지(기울기, Gradient)를 계산합니다.  

    2. 경사 하강법: 계산된 기울기를 사용하여 손실이 최소화되는 방향으로 가중치와 편향을 업데이트합니다.  

$$\text{새로운 } W = \text{기존 } W - \text{학습률} \cdot \frac{\partial \text{Loss}}{\partial W}$$

3.4. 최적화 도구 (Optimizer)

경사 하강법을 효율적으로 수행하기 위한 알고리즘입니다. 학습률을 동적으로 조절하거나 과거의 기울기 정보를 활용하여 **최적해(Minimum Loss)**를 더 빠르고 안정적으로 찾도록 돕습니다. (예: $\text{SGD}$, $\text{Momentum}$, $\text{Adam}$ 등)

4. 주요 신경망 아키텍처

아키텍처 | 설명 | 주요 용도
|--|--|--|
MLP / DNN | 다층 퍼셉트론 / 심층 신경망. 가장 기본적인 신경망 형태. 각 층의 뉴런들이 다음 층의 모든 뉴런과 연결됨 (Dense/Fully Connected Layer). | 단순 분류/회귀, 정형 데이터 분석.
CNN | 합성곱 신경망 (Convolutional Neural Network). 합성곱 필터(Convolutional Filter)를 사용하여 공간적 특징을 추출. | 이미지 처리, 컴퓨터 비전. 
RNN / LSTM | 순환 신경망 (Recurrent Neural Network) / 장단기 메모리 (Long Short-Term Memory). 시간의 흐름에 따라 순서가 있는 데이터를 처리하기 위해 이전 시점의 정보를 기억하며 순환 구조를 가짐. | 시계열 데이터, 자연어 처리(번역, 텍스트 생성).
