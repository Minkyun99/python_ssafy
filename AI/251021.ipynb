# ⭐ Pre-training VS Post-training

### 거대 언어 모델 학습 패러다임
1. 최근 거대 언어 모델을 학습하기 위해서는 크게 두 가지 단계로 나눠짐
    - Pre-training(사전학습) : 방대한 인터넷 텍스트를 통해 언어와 지식을 배우는 단계
    - Post-training(사후학습) : 사람이 원하는 방식으로 대화하고, 안전하고, 유용하게 만드는 단계

2. Pre-training(사전학습)
    - 방대한 인터넷 텍스트 데이터를 이용한 Self-supervised learning을 통해 언어 패턴, 지식 등을 배움
    - 학습 목표 : 다음 단어 예측 (Next Token Prediction)
    - 예시 : "내일은 비가 _____" ➡️ "온다" 의 확률을 가장 높게 만들도록 파라미터 업데이트  
    ![image.png](./img/ai145.png)  

2. Pre-training 후 모델의 응답 예시
    - 다음 단어를 예측하는 데에 강점을 보이지만 질문에 대한 대답을 하지는 않음  
    ![image.png](./img/ai146.png)  

3. Pre-training(사후학습)
    - 유저의 의도를 파악하고 원하는 답변을 모델이 응답하도록 사후 학습을 진행함
    - 대표 기법
        - Instruction-tuning, RLHF(Reinforcenemt Learning from Human Feedback), DPO(Direct Preference Optimization), RLVR(Reinforcement Learning with Verifiable Reward)

### ReinForcemen Learning from Human Feedback(RLHF)
1. Instruction-tuning의 한계
    - 명확한 한계점 중 하나는 "정답 데이터를 수집하는 것이 비싸다" 인데, 명확한 한계점 말고 다른 문제들은 뭐가 있을까?
    - 문제 1 : 개방형/창의적 생성과 같은 태스크에는 정답이 존재하지 않음
        - EX) 개와 개가 기르는 메뚜기에 대한 이야기를 만들어줘
    - 문제 2 : 언어 모델링은 모든 토큰 레벨의 오류를 동일하게 취급하지만, 어떤 오류는 다른 오류보다 심하게 작용함
    - 문제 3 : 사람이 만든 답변(정답 레이블)이 최적이 아닐 수 있음

2. Instruction-tuning의 한계
    - Instruction-tuning을 하더라도, 언어모델의 목표(LM objective)와 "인간의 선호를 만족시키는 것" 사이에는 여전히 불일치가 존재함
    - 그렇다면, 인간의 선호(Human Preference)를 만족시키기 위해서 어떻게 해야 할까?

3. 인간의 선호도(Human Preference)란?
    - 단답형 질문
        - Q : 상대성 이론을 발표한 과학자는 누구인가?
        - A : 알베르트 아인슈타인
    - 주관식 질문
        - Q : 아인슈타인이 오늘날 살아있다면, 현대 과학기술에 어떤 기여를 했을까?  
        ![image.png](./img/ai154.png)  

4. 인간의 선호를 반영한 최적화(Optimizing for human preferences)
    - 언어모델을 학습한다고 가정 했을 때(e.g., 요약 태스크)
    - 어떤 지시문(instruction) x와 언어모델의 응답 y가 있을 때, 해당 요약에 대한 인간이 준 보상(human reward)을 얻을 수 있다고 상상해보자 : R(x, y) \\in R, 점수가 높을수록 좋음  
    ![image.png](./img/ai156.png)  
    - 이제 우리는 언어모델의 응답 중 기대 보상(expected reward)를 최대화 하는 것을 목표로 함
    $$\text{E}_{\hat{y} \sim p_{\theta}(y|x)}[R(x, \hat{y})]$$

# ⭐ 거대 언어 모델 (LLM) 학습의 3단계

    거대 언어 모델($\text{LLM}$)의 개발 과정은 방대한 데이터로 일반 지식을 쌓는 것부터 시작하여, 특정 목적에 맞게 모델을 다듬어가는 일련의 체계적인 과정입니다.

단계 | 목적 | 사용 데이터 | 결과 모델
|--|--|--|--|
1.사전 학습 (Pre-training) | 언어의 구조, 통계적 패턴, 방대한 일반 지식 습득 | 대규모 웹 크롤링 데이터, 도서, 위키피디아 등 (수백 테라바이트) | 기초 모델 (Foundation Model)
2.사후 학습 (Post-training) | 인간의 선호도와 안전 기준에 맞춰 모델의 출력 품질과 안전성을 향상 | 고품질의 소규모 데이터, 인간 평가자가 레이블링한 비교 데이터 | 정렬된 모델 (Aligned Model)
3.파인 튜닝 (Fine-tuning) | 특정 태스크나 도메인에 모델을 특화시키거나, 서비스용 대화체로 전환 | 특정 태스크 데이터셋 (질의응답, 분류) 또는 명령어 데이터셋 (Instruction Data) | 특화/서비스 모델

1. 사전 학습 (Pre-training)

    사전 학습은 $\text{LLM}$ 개발의 가장 첫 단계이자 비용이 많이 드는 단계입니다. 이 단계에서 모델은 언어와 세상에 대한 기초 지식을 습득합니다.

    - 주요 목표: 대규모 텍스트 데이터셋을 학습하여 언어의 통계적 패턴, 문법, 광범위한 사실적 지식을 습득하는 것입니다. 이 과정을 통해 **기초 모델(Foundation Model)**이 탄생합니다.

    - 학습 방법:

        - 다음 토큰 예측 (Next Token Prediction, $\text{Causal Language Modeling}$): (주로 $\text{GPT}$ 계열의 $\text{Decoder}$ 모델) 이전 단어들을 보고 다음에 올 단어를 예측하도록 학습합니다. 이를 통해 문맥 이해 및 자연스러운 문장 생성 능력을 갖춥니다.

        - 마스크 언어 모델링 (Masked Language Modeling, $\text{MLM}$): (주로 $\text{BERT}$ 계열의 $\text{Encoder}$ 모델) 문장에서 일부 단어를 가리고, 가려진 단어를 예측하도록 학습합니다. 문맥 파악 능력을 강화합니다.

    - 결과: 세상의 일반 지식을 학습한 기초 모델이 생성됩니다. 이 모델은 지식은 풍부하지만, 아직 사용자의 지시에 정확히 따르거나 안전하게 대화하는 능력은 부족합니다.

2. 사후 학습 (Post-training)

    사전 학습을 마친 모델을 인간의 가치관, 선호도, 그리고 안전 기준에 맞게 조정하는 과정입니다. 이를 **모델 정렬(Alignment)**이라고도 합니다.

    - 주요 목표: 모델이 유용하고(Helpful), 정직하며(Harmless), 무해하도록(Honest) 만듭니다. 모델이 지시를 더 잘 이해하고, 비윤리적이거나 유해한 콘텐츠 생성을 회피하도록 훈련합니다.

    - 대표적인 방법: $\text{RLHF}$ ($\text{Reinforcement Learning from Human Feedback}$)

        - 지도 파인 튜닝 ($\text{SFT}$, $\text{Supervised Fine-Tuning}$): 모델에게 고품질의 지시-응답 쌍 데이터를 학습시켜 지시를 따르는 기본 능력을 향상시킵니다.

        - 보상 모델 학습 ($\text{Reward Model}$): 인간 평가자가 모델의 여러 응답 중 가장 선호하는 순서를 매긴 데이터를 수집하고, 이를 학습시켜 인간의 선호도를 예측하는 보상 모델을 만듭니다.

        - 강화 학습 ($\text{RL}$): 이 보상 모델을 사용하여 $\text{LLM}$을 강화 학습시킵니다. 모델은 보상 점수를 최대화하는 방향으로 응답을 생성하도록 훈련받습니다.

    - 결과: 인간의 가치에 맞춰 정렬되어, 안전하고 유용한 대화를 수행할 수 있는 모델이 됩니다.

3. 파인 튜닝 (Fine-tuning)

    파인 튜닝은 기초 모델이나 사후 학습을 거친 모델을 특정 작업이나 도메인에 특화시키는 과정입니다.

    - 주요 목표:

        - 특정 태스크 최적화: 감성 분석, 개체명 인식, 기계 번역 등 특정 $\text{NLP}$ 태스크에서 최고의 성능을 내도록 훈련합니다.

        - 도메인 특화: 법률, 의학, 금융 등 특정 전문 분야의 용어와 문맥을 깊이 있게 이해하고 활용하도록 훈련합니다.

        - 명령어 파인 튜닝 ($\text{Instruction Fine-Tuning}$): 다양한 형태의 명령어와 그에 대한 올바른 응답 쌍을 학습시켜, 사용자의 복잡한 지시를 더욱 정확하게 이해하고 따르는 범용 대화 모델로 만듭니다. (일반적으로 사후 학습의 첫 단계로 사용되기도 합니다.)

    - 결과: 특정 업무나 도메인에 특화되어, 해당 분야에서 높은 정확도와 성능을 보이는 실제 서비스용 모델이 탄생합니다.

# ⭐ K-LLMU (Korean Language Model Utility) 벤치마크 설명

K-LLMU는 한국어 기반 대규모 언어 모델(LLM)의 능력과 유용성(Utility)을 종합적으로 측정하기 위해 설계된 평가 체계입니다. 기존의 범용 벤치마크(예: MMLU, SuperGLUE)가 한국어와 한국 문화의 특성을 반영하지 못하는 한계를 극복하기 위해 만들어졌습니다.

1. 개발 배경 및 목적

항목 | 내용 
|--|--|
필요성 | 글로벌 벤치마크는 주로 영어 기반이며, 한국어 고유의 문법, 관용 표현, 문화적 지식을 평가하는 데 한계가 있습니다.
목표 | 국내외에서 개발되는 LLM들이 한국어 환경에서 얼마나 정확하고 유용하며 안전하게 작동하는지 객관적으로 측정하는 표준을 제시합니다.
의의 | 한국어 LLM의 성능 향상과 투명한 비교를 유도하여 국내 AI 생태계 발전에 기여합니다.

2. 주요 평가 영역 및 특징

K-LLMU는 모델의 단순한 언어 이해 능력을 넘어, 한국 사회와 관련된 심화된 능력까지 평가하는 것이 특징입니다. 평가 영역은 일반적으로 모델의 성능(Competency)과 안전성(Safety) 측면을 모두 포함합니다.

A. 언어 능력 및 지식 (Competency)

- 한국어 추론: 복잡한 한국어 문장 구조나 논리적 관계를 얼마나 정확하게 파악하고 추론하는지 평가합니다.

- 한국 문화/역사 지식: 한국의 특정 역사적 사건, 인물, 고유 문화 등에 대한 이해도를 평가하여, 모델이 단순 번역을 넘어 문맥을 이해하는지 확인합니다.

- 법률/윤리적 판단: 한국의 법률 체계나 사회적 규범에 기반한 윤리적 질문에 대해 적절하게 판단하는 능력을 측정합니다.

- 전문 분야 지식: 한국어로 된 금융, 의학, 과학 등 전문 분야의 질문에 대한 답변 정확도를 평가합니다.

B. 유용성 및 안전성 (Utility & Safety)

- 안전성 및 유해성: 한국어로 유해하거나 차별적인 콘텐츠, 또는 불법적인 정보를 생성하려는 시도에 대해 모델이 얼마나 잘 방어하는지를 측정합니다.

- 편향성: 성별, 지역, 정치적 성향 등에 대한 특정 편향을 드러내지 않고 중립적이고 공정한 응답을 하는지 평가합니다.

3. K-LLMU가 중요한 이유

K-LLMU는 한국어 모델 개발자들이 목표로 삼아야 할 구체적인 기준점을 제공합니다. 이 벤치마크에서 높은 점수를 얻는 모델은 한국 시장에서 더욱 신뢰성 있고 실용적인 서비스 제공이 가능하다고 인정받게 됩니다.

# 🤖 인간 개입형 머신러닝 (HITL) 순환 구조 정리
![image.png](./img/ai155.png)  

**인간 개입형 머신러닝($\text{Human-in-the-loop ML}$)**은 인공지능 모델의 학습, 평가, 개선 과정에 인간 전문가의 판단을 필수적으로 포함시키는 체계적인 방법론입니다. 이 순환 구조는 모델이 스스로 해결하기 어려운 문제(불확실성)에 대해서만 인간의 지능을 활용하여 효율성과 정확도를 동시에 극대화합니다.

1. HITL 순환의 4단계 핵심 키워드

단계 | 키워드 | 역할 및 목적
|--|--|--|
1단계 | Labeled Data | 모델 학습의 기반이 되는 정답(레이블)이 부여된 고품질 데이터.
2단계 | Training Model | 레이블된 데이터를 사용하여 모델을 훈련하고 패턴을 학습시킵니다.
3단계 | Selecting Confusing Data | 훈련된 모델이 예측 신뢰도가 낮거나, 예외적(Edge Cases)이라 판단하여 인간의 도움이 필요하다고 선별한 데이터를 추출합니다.
4단계 | Human Annotator | 선별된 데이터에 대해 **정확한 정답 레이블 또는 새로운 지침(Instruction)**을 부여하는 인간 전문가.

2. 순환 구조의 원리 (피드백 루프)

    $\text{HITL}$의 핵심은 **3단계(Selecting Confusing Data)** 에서 시작된 데이터가 **4단계(Human Annotator)**를 거쳐 다시 **1단계(Labeled Data)** 로 피드백되는 지속적인 개선 루프를 만든다는 것입니다.

    1. AI의 한계 인식: 모델은 자신이 혼란스러운 데이터를 스스로 식별합니다.

    2. HI의 개입: 인간은 $\text{AI}$가 판단하지 못한 부분에 대해 **가장 가치 있는 정보(정답)** 를 제공합니다.

    3. 모델 개선: 이 새롭게 레이블링된 고품질 데이터로 모델을 재학습(파인 튜닝)하여, 이전에 틀렸던 영역에 대한 지식을 빠르게 보강합니다.

3. HITL의 필요성

    - 정확도 극대화: 모델이 가장 취약한 부분에 대해서만 집중적으로 학습하여 전반적인 성능을 높입니다.

    - 비용 효율성: 모든 데이터를 인간이 레이블링하는 대신, 필요한 데이터만 선별하여 인력 및 시간 비용을 절감합니다.

    - 안전성 및 편향성 관리: $\text{LLM}$의 **사후 학습($\text{Post-training}$)** 과정에서 인간의 윤리적 기준과 가치관을 반영하여 안전한 출력을 보장하는 데 필수적으로 활용됩니다.
