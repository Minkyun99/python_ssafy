# ⭐ AI 수업 핵심 정리 (DAY1)

1. 🧑‍🏫 AI, ML, DL의 정의 및 기초

개념 | 정의 | 예시
|--|--|--|
AI (Artificial Intelligence) | 주어진 환경/데이터를 인지, 학습, 추론하여 목표 달성을 예측, 행동하는 시스템. | 규칙 기반 시스템, 휴리스틱 기반 알고리즘 (ML이 아닌 AI의 예)
ML (Machine Learning) | AI 범주 내에서 데이터로부터 학습하여 목적을 달성하는 접근 방법론. | 언어 모델, 이미지 분류 모델, 추천 시스템
DL (Deep Learning) | ML 범주 내에서 신경망(Neural Network) 함수를 사용한 학습 방법론. | -

**데이터 구성 요소**

- Feature (피처, 특성): 모델이 예측에 사용하는 입력 정보 (예측의 근거/단서).
- Label (라벨, 목표값): 모델이 예측하려는 정답 (학습의 목표값).

모델과 학습

- 가설 공간 (Hypothesis Space): 입력 $\rightarrow$ 출력 관계를 표현할 수 있는 모든 후보 함수들의 모음.
- 모델 (Model): 가설 공간 $\text{F}$에 속한 특정 함수 $\text{f}$.
- 학습 (Learning): 주어진 데이터와 성능 척도를 바탕으로 가설 공간 $\text{F}$의 후보들 중 최적의 모델을 선택하는 과정.
- 일반적 모델 가정: $\text{Y} = \text{f}^{*}(\text{X}) + \text{e}$ (측정 오차 $\text{e}$는 피쳐 $\text{X}$와 독립, $\text{E}[\text{e}] = 0$ 가정).

2. 지도 학습 (Supervised Learning)

입력($\text{특성}$)과 정답($\text{라벨}$)이 쌍으로 있는 데이터로, 새 입력에 대한 정답을 잘 맞추는 규칙을 학습하는 것이 목표입니다.

2-1. 회귀 문제 (Regression)

입력으로부터 숫자 (가격, 점수 등)를 예측합니다.

지표 | 공식 | 해석
|--|--|--|
평균제곱오차 ($\text{MSE}$) |$\frac{1}{n}\sum_{i=1}^{n} (\text{y}_i - \hat{\text{y}}_i)^2$ |각 데이터에서 정답과 예측의 평균 제곱 차이값 (큰 오류에 더 큰 벌칙).
$\text{R}^2$ (결정계수)|$1 - \frac{\sum_{i=1}^{n} (\text{y}_i - \hat{\text{y}}_i)^2}{\sum_{i=1}^{n} (\text{y}_i - \bar{\text{y}})^2}$|라벨의 분산 중 특성으로 설명되는 비율 ($\text{0} \sim \text{1}$ 사이). $\text{1}$에 가까울수록 설명력이 높음. (음수도 나올 수 있음).

2-2. 분류 문제 (Classification)

입력으로부터 범주 (스팸/정상, 악성/양성 등)를 가려냅니다.

**혼동 행렬 (Confusion Matrix) 기반 지표**

지표 | 공식 | 정의 및 특징
|--|--|--|
정확도 (Accuracy) |$\frac{\text{TP} + \text{TN}}{\text{TP} + \text{TN} + \text{FP} + \text{FN}}$|전체 중 맞춘 비율. 데이터가 불균형할 경우 안전하지 않음.
정밀도 (Precision) |$\frac{\text{TP}}{\text{TP} + \text{FP}}$|양성이라 판정한 것 중 진짜 양성의 비율.
재현율 (Recall) |$\frac{\text{TP}}{\text{TP} + \text{FN}}$|진짜 양성 가운데 잡아낸 예측 양성의 비율.
F1-score |$2 \cdot \frac{\text{Precision} \cdot \text{Recall}}{\text{Precision} + \text{Recall}}$|정밀도와 재현율의 조화 평균.

3. 모델의 일반화 능력

3-1. 과적합 (Overfitting) vs. 과소적합 (Underfitting)

- 과적합 (Overfitting): 훈련 데이터의 우연한 패턴/잡음까지 외워버려 (모델이 너무 복잡), 훈련에서는 잘 맞지만 테스트 성능이 나빠지는 현상 (일반화 실패).

- 과소적합 (Underfitting): 모델이 너무 단순하거나 학습이 불충분하여 중요한 패턴을 놓치는 현상 (오류 큼).

- 해결 실마리: 더 많은 데이터, 교차 검증을 통한 모델 선정.

3-2. 테스트 성능 평가

- 훈련 오류 (Training Error): 모델을 학습시킨 같은 데이터에 대한 오류. (테스트 오류를 과소평가함).

- 테스트 오류 (Test Error): 학습에 쓰지 않은 새 관측치에 대한 평균 예측 오류. 모델의 진짜 성능을 판단하는 기준.

- 테스트 오류 추정 대안:
    - 검증셋 (Validation Set) 접근: 샘플을 훈련셋과 검증셋으로 분할하여 검증셋 오류를 계산.
        - 단점: 훈련셋 규모 축소로 인해 전체 데이터로 학습했을 때보다 성능을 과대 추정할 수 있음.

- K-겹 교차검증 (K-fold Cross-Validation): 데이터를 $\text{K}$개 폴드로 분할하여, $\text{k}$번째 폴드를 검증셋으로 사용하는 과정을 $\text{K}$번 반복한 후 평균 오류로 테스트 오류를 추정. (테스트 오류 추정의 표준적 접근).

- Leave-One-Out 교차 검증 ($\text{LOOCV}$): $\text{n}$번 반복하여 한 관측치만 검증셋으로 사용하는 극단적인 $\text{K}$-겹 교차검증 ($\text{K} = \text{n}$).

4. 비지도 학습 (Unsupervised Learning) - 클러스터링

레이블($\text{정답}$) 없이 데이터의 구조, 패턴, 잠재 서브그룹을 찾아내는 학습입니다.

과제 | 목표
|--|--|
군집화 (Clustering) | 서로 비슷한 데이터끼리 묶어 동질 그룹 만들기 (예: 고객 세그먼테이션).
차원 축소 | 데이터의 복잡성을 줄여 시각화나 효율적인 학습을 돕습니다.

**클러스터링 기법 비교**

구분 | K-means 클러스터링 | 계층적 군집 (Hierarchical Clustering)
|--|--|--|
군집 구조 | 평면적 (Flat) | 계층적 (Hierarchical) - 덴드로그램 제공
K 값 지정 | 미리 지정해야 함 ($\text{K}$-평균) | $\text{K}$를 고정하지 않고 구조 전체를 제공
핵심 아이디어 | 클러스터 내부 변동의 합을 최소화하는 분할을 찾음. | 가장 가까운 클러스터들을 순차적으로 병합 (상향식). | 
알고리즘 특성 | 초기값에 따라 지역 최솟값으로 수렴 가능. 여러 번 시도 권장. | 매 단계 클러스터 쌍 간의 거리 계산 필요. 계산량이 많음. 

**계층적 군집의 링크(Link) 유형**

군집 간의 거리를 정의하는 방법이며, 어떤 링크를 선택하느냐에 따라 군집 결과가 달라집니다.

링크 유형 | 거리 계산 기준 | 특징
|--|--|--|
Single Link | 두 군집 내 데이터 쌍별 거리 중 최소값 | 연쇄적 구조, 노이즈에 약함 (길고 가늘게 뭉침).
Complete Link | 두 군집 내 데이터 쌍별 거리 중 최대값 |조밀, 이상치에 강함 (둥글고 조밀하게 뭉침).
Average Link | 두 군집 내 데이터 쌍별 거리의 평균 | 균형 잡히고 안정적 (중간 형태).

**❗ 클러스터링 시 주의점**

- 스케일링: 변수 단위 차이의 영향을 줄이기 위해 **표준화(Scaling)** 가 필요할 수 있음. 

- $\text{K}$ 개수 선정: 적절한 클러스터 개수($\text{K}$)를 정하는 것은 어려운 문제이며, 합의된 정답은 없음.

3. 로지스틱 회귀 (Logistic Regression)

3-1. 로지스틱 함수 (Sigmoid)

- 목표: 선형 함수의 결과($\text{z}$)를 $\text{0}$과 $\text{1}$ 사이의 확률로 변환합니다.

- 함수: $\text{g(z)} = \frac{1}{1 + e^{-z}}$

- 특징: 이진 분류 문제에서 $1$이 될 확률 ($\text{P}(\text{y}=1|\text{x})$)을 추정합니다.

3-2. 모델 학습 목표: 우도 (Likelihood)

로지스틱 회귀는 확률을 추정하는 모델이므로, 회귀에서 사용했던 $\text{MSE}$ 대신 **우도($\text{Likelihood}$)** 를 사용하여 모델의 적합도를 평가합니다.

개념 | 정의 | 학습 목표
|--|--|--|
우도 ($\text{L}$) | **"현재 확률 함수가 데이터를 얼마나 잘 설명하는지"**를 나타내는 지표. | 우도 값을 최대화하도록 모수 ($\text{B}$)를 추정함 ($\text{MLE}$).
$\text{MLE}$ | Maximum Likelihood Estimation (최대 우도 추정). | 우도 ($\text{L}(\text{B})$)를 최대화하는 $\text{B}$ (계수/모수)를 찾습니다.

3-3. 최적화 과정

1. 로그 우도 ($\text{Log-Likelihood}$) 변환:

- 우도 함수는 곱셈으로 이루어져 있어 미분이 복잡합니다.

- 양변에 $\text{log}$를 취해 곱셈을 덧셈으로 변환하여 **로그 우도** ($\text{log L}(\text{B})$)를 만듭니다.

2. 수치적 최적화:

- $\text{log L}(\text{B})$를 $\text{B}$에 대해 미분하여 **도함수 = 0** 이 되는 지점을 찾습니다.

- 선형 회귀처럼 공식이 아닌, 경사 하강법과 같은 수치적(반복) 최적화를 통해 $\text{B}$를 찾아 나갑니다.

2. 🧑‍🏫 순환 신경망 (RNN) 및 장기 의존성 문제 해결

2-1. 순환 신경망 (RNN)

- 핵심: 이전 단계의 은닉 상태($\text{Hidden State}$)를 현재 단계의 입력으로 사용하여 순서/문맥 정보를 처리함.

- 구조: $\text{Recurrent}$ 루프를 통해 시퀀스를 반복적으로 처리.

- 한계: 장기 의존성 문제 (Long-Term Dependency) 발생. 시퀀스가 길어질수록 초반의 정보가 사라져 뒷부분에 영향을 미치지 못함.

2-2. 장기-단기 메모리 (LSTM)

- 목표: $\text{RNN}$의 장기 의존성 문제 해결.

- 핵심 아이디어: **셀 상태 ($\text{Cell State}$)**라는 별도의 메모리 경로를 추가하고, $\text{Gate}$ 메커니즘을 통해 정보의 흐름을 제어.

게이트 (Gate) | 역할
|--|--|
망각 게이트 ($\text{Forget Gate}$) | 과거 정보 중 잊어버릴 정보를 결정 ($\text{0}$에 가까울수록 망각).
입력 게이트 ($\text{Input Gate}$) | 현재 입력 중 새로 저장할 정보를 결정.
출력 게이트 ($\text{Output Gate}$) | 셀 상태를 기반으로 현재 은닉 상태 ($\text{h}_{\text{t}}$)로 출력할 내용을 결정.

3. 🧑‍🏫 트랜스포머 모델 (Transformer)의 구조와 활용

3-1. 트랜스포머의 구조적 특징

모델 유형 | 설명 | 주요 사용 태스크
|--|--|--|
Encoder 모델 | 입력 시퀀스 전체를 한 번에 읽고 **문맥적 표현(Contextual Representation)** 을 추출. | 분류, 문장 유사도 판단, 마스킹된 단어 예측 ($\text{BERT}$)
Decoder 모델 | 이전 토큰들을 기반으로 다음 토큰을 순차적으로 생성 ($\text{Autoregressive}$). | 텍스트 생성, 대화, 번역 ($\text{GPT}$)
Encoder-Decoder | 입력 인코딩 후, 이를 기반으로 디코딩하여 출력 시퀀스 생성. | 기계 번역, 요약

3-2. Decoder 모델의 부상

- 현 트렌드: 최근에는 $\text{Decoder}$만 사용하는 $\text{LLM}$이 대세 (예: $\text{GPT}$ 계열).

- 이유: $\text{Decoder}$의 성능이 $\text{Encoder}$보다 좋기 때문에, $\text{Encoder}$를 개발할 자원을 $\text{Decoder}$ 확장에 사용하여 모델 성능을 극대화할 수 있음.

3-3. Decoder 모델 활용

태스크 | 활용 방식
|--|--|
생성 태스크 (대화, 요약) | 사전학습과 동일하게 다음 단어 예측 방식으로 $\text{Fine-tuning}$을 수행.   (시퀀스 출력에 자연스럽게 적합)
분류 태스크 | 마지막 은닉 상태($\text{Hidden State}$) 위에 새로운 선형 레이어($\text{Classifier}$)를 연결하여 사용.  ($\text{Fine-tuning}$ 시 전체 $\text{Decoder}$에 $\text{Gradient}$ 전파)

4. 🧑‍🏫 인-컨텍스트 학습 (In-Context Learning, ICL)

- 개념: $\text{GPT-3}$에서 처음 도입. 모델의 **가중치($\text{Weights}$)를 업데이트하지 않고** (학습 없이), 프롬프트 내에 몇 가지 **예시(Few-shot)** 를 제공하는 것만으로 새로운 태스크를 수행하는 능력.

- Zero-shot: 예시 없이 태스크 설명만 제공.

- Few-shot: 태스크 설명과 함께 몇 가지 $\text{Input/Output}$ 예시를 제공.

- 원리: 모델이 내부적으로 가진 거대한 사전지식을 활용하여, 예시를 통해 태스크의 형식과 패턴을 파악하고 결과를 생성함.

2. 🧑‍🏫 멀티모달 AI (Vision-Language Model, VLM)

2-1. VLM의 기본 개념

- 멀티모달 AI: 텍스트, 이미지, 비디오 등 **서로 다른 양식(Modality)**의 데이터를 동시에 처리하고 이해하는 AI.

- 핵심 목표: 이미지와 텍스트 간의 의미론적 정렬(Semantic Alignment). 즉, "이 이미지가 이 텍스트와 일치한다"는 것을 학습.

- 주요 태스크: 이미지 캡셔닝 (Image Captioning), 시각적 질의응답 (VQA), 멀티모달 생성.

2-2. CLIP (Contrastive Language-Image Pre-training) 모델

특징  | 설명 | 학습 방식 | 주요 활용
|--|--|--|--|
대규모 데이터 | 대규모 웹 이미지-텍스트 쌍으로 학습. | 대조 학습 (Contrastive Learning): 올바른 이미지-텍스트 쌍의 유사도는 높이고, 잘못된 쌍의 유사도는 낮추도록 학습. | Zero-shot 분류: 학습 시 보지 못한 클래스도 분류 가능.
구조 | 이미지 인코더와 텍스트 인코더가 별도로 존재하며, 이들의 임베딩 공간을 정렬함. | 시각 인코더와 어댑터 중심의 최적화를 통해 시각-언어 대응 학습. | -
