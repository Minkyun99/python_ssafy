##### Pandas 패키지의 **Series**에 대해 자세히 정리해 드립니다. 
- $\text{Series}$는 $\text{Pandas}$에서 데이터를 다루는 데 가장 기본이 되는 1차원 배열(Array) 구조입니다.


### 🐍 Pandas Series 개요
**Series**는 Pandas 라이브러리의 핵심 데이터 구조 중 하나로, 1차원 배열과 유사하며, 각 데이터 요소에 인덱스(Index) 레이블이 붙어 있는 구조입니다. $\text{NumPy}$의 1차원 배열과 비슷하지만, 인덱스에 이름을 부여하거나 문자열을 사용할 수 있다는 점에서 차이가 있습니다.

1. Series의 특징 및 구성 요소
$\text{Series}$는 크게 두 부분으로 구성됩니다.

구성 요소|	설명	|Python 객체
|--|--|--|
값 (Values)|	실제 저장되는 데이터입니다. $\text{NumPy}$의 `ndarray` 형태로 저장되며, 동일한 \*\*데이터 타입($\text{dtype}$)**을 가집니다.	|series.values
인덱스 (Index)|	각 값에 대응하는 라벨입니다. 기본적으로 0부터 시작하는 정수 인덱스가 부여되지만, 사용자가 임의의 이름(문자열, 날짜 등)으로 정의할 수 있습니다.|	series.index


```python
# 2. Series 생성 방법
# $\text{Series}$는 주로 리스트, NumPy 배열, 또는 딕셔너리를 사용하여 생성합니다.

import pandas as pd
import numpy as np

# 1. 리스트로 Series 생성 (인덱스는 0부터 시작)
s1 = pd.Series([10, 20, 30, 40])
# Index: 0, 1, 2, 3
# Values: 10, 20, 30, 40

# 2. 인덱스를 명시하여 Series 생성
s2 = pd.Series([100, 200, 300], index=['a', 'b', 'c'])
# Index: 'a', 'b', 'c'
# Values: 100, 200, 300

# 3. 딕셔너리로 Series 생성 (Key가 Index가 됨)
data = {'서울': 990, '부산': 340, '대구': 240}
s3 = pd.Series(data)
# Index: '서울', '부산', '대구'
```

3. Series의 주요 기능 및 메서드
$\text{Series}$는 데이터 분석을 위해 NumPy 배열의 모든 기능을 지원할 뿐만 아니라, 레이블 기반의 인덱싱 및 다양한 통계 메서드를 제공합니다.

A. 데이터 접근 (인덱싱 및 슬라이싱)
$\text{Series}$는 위치 기반(iloc)과 라벨 기반(loc) 두 가지 방식으로 데이터를 접근할 수 있습니다.

메서드/방식|	설명	|예시
|--|--|--|
라벨 인덱싱|	인덱스 이름(라벨)을 사용하여 값을 가져옵니다.	|s2['a'] -> 100
위치 인덱싱|	정수 위치(0부터 시작)를 사용하여 값을 가져옵니다.	|s2.iloc[0] -> 100
슬라이싱|	$\text{NumPy}$와 동일하게 위치 기반으로 슬라이싱할 수 있습니다.	|s1[1:3] -> [20, 30]


B. 통계 및 기술 분석 메서드
EDA(탐색적 데이터 분석)에 필수적인 통계 메서드를 제공합니다.

메서드	|설명	|예시 (결과)
|--|--|--|
s.mean()|	평균값 계산	|25.0
s.std()|	표준편차 계산	|(값)
s.min(), s.max()|	최소값, 최대값|	(값)
s.describe()	|개수, 평균, 표준편차, 4분위수 등 요약 통계량 반환	|Series
s.value_counts()	|고유 값별 빈도수 계산 (범주형 데이터 분석)|	Series
s.unique()	|고유한 값들만 NumPy 배열로 반환	|array([10,20,30,40])
s.kurt()	|첨도(Kurtosis) 계산	|(값)
s.skew()	|왜도(Skewness) 계산	|(값)

C. 연산 및 브로드캐스팅
$\text{Series}$는 NumPy 배열처럼 브로드캐스팅(Broadcasting)을 지원하여 벡터 연산을 효율적으로 수행합니다.

- 스칼라 연산: s1 * 2 (모든 요소에 2를 곱함)
- 벡터 연산: s1 + s1 (두 $\text{Series}$를 요소별로 더함)

4. DataFrame과의 관계
$\text{DataFrame}$은 Series 객체들이 모여서 만들어진 2차원 구조입니다.

구조	|차원	|구성 단위
|--|--|--|
Series	|1차원	|값 + 인덱스
DataFrame	|2차원	|$\text{Series}$들의 집합 (각 열이 Series)

1. 분류의 정의
- 분류 : 정해진 범주(카테고리) 중 하나로 지정하는 것

2. 분류 함수의 목표
- 입력(특성 벡터) : 눈동자 이미지, 이메일 내용/제목

##### 🧑‍🏫 MLE 활용 모수 추정
1. 우도(Likelihood)
- 확률을 추정하는 모델을 결정했으니, 모수를 추정하는 방법에 대해 알아야 함
- 선형회귀에서 '현재 함수가 데이터와 오차가 작은지'를 평가하기 위해 평균 제곱 오차(MSE)를 지표로 삼았듯, 확률을 계산하는 함수를 평가하기 위해선 우도(Likelihood)를 지표로 삼음
- 우도란, '현재 확률 함수가 데이터를 얼마나 잘 설명하는지를' 나타낸 지표임
- 그러므로, 모델의 학습은 우도 값을 높여 최대화가 되도록 하는 것이 목표이며, 이를 Maximum Likelihood Estimation(MLE)이라 함

2. 로지스틱 회귀와 같은 우도 최대화
- 로지스틱 회귀와 같은 이진 분류 문제에서 우도를 최대화함  
![image.png](./img/ai54.png)  
- 하지만, 위와 같은 곱으로 이뤄진 함수의 경우, B에 대해 미분이 어렵기 때문에 양 변에 log를 취해 곱셈을 더하기로 변환시킨 뒤 log-likelihood를 만들어 최대화함  
![image.png](./img/ai55.png) 
- log L(B)를 미분하여 **도함수=0**에 근접하도록 수치적(반복) 최적화를 통해 B들을 찾아 감

**❗우도를 구하는 이유**
- y 값이 1에 가까운 값은 최대한 1에 가깝게 만들어주는 것이 좋음
- y 값이 1에 가까울수록 현재 관측된 데이터가 얼마나 잘 설명하는지를 보여주기 때문
- 반대로 y 값이 0에 가까운 값은 최대한 0에 가깝게 만들어주는 것이 좋음

3. log-likelihood로 변형하여 최대화하는 B를 구해도 되는가?
- 로지스틱 회귀에서 우도(likelihood)를 최대화하는 것인데, log 변환을 한 log-likelihood를 최대화 하는 것으로 변형하여 문제를 접근해도 과연 괜찮을까?
- Log 함수는 단소(monotone) 증가 함수임
- 따라서, log-likelihood를 최대화하는 B와 likelihood를 최대화하는 B는 같음
