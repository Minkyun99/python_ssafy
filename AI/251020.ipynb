# ⭐ AI 학습 방법론

### AI 학습 방법론: 정렬, 지시, 선호 학습 비교 분석

최근 $\text{LLM}$ (거대 언어 모델)의 성능 향상은 이 세 가지 후처리 학습 단계에 크게 의존합니다. 이들은 모델이 '무엇을 할 수 있는지'를 넘어서 '무엇을 해야 하는지'를 배우도록 돕습니다.

1. 🧑‍🏫 지시 학습 (Instruction Tuning)

지시 학습은 모델에게 특정 지침(Instruction)에 따라 응답하는 능력을 가르치는 과정입니다.

구분 | 내용 
|--|--|
목표 | 모델이 사용자의 구체적인 질문, 명령, 요구사항을 이해하고 정확하게 따르게 만듭니다.
방법 | (지시/프롬프트, 원하는 응답) 쌍으로 구성된 대규모 데이터셋을 사용하여 모델을 파인 튜닝(Fine-tuning)합니다.
효과 | 모델의 일반화 능력을 향상시켜, 학습 데이터에서 보지 못한 새로운 형태의 지시에도 일관성 있고 유용한 응답을 생성할 수 있게 합니다.
예시 | "이 문단을 세 문장으로 요약해 줘." 또는 "셰익스피어 스타일로 시를 써 줘."와 같은 명령에 대응하는 훈련.

⭕ **지시 학습의 역할**

지시 학습은 모델이 'Helpful'(도움이 되는)하게 만드는 첫 번째 단계입니다. 이 과정이 없으면, 모델은 단순히 훈련 데이터에서 가장 확률 높은 다음 단어를 예측할 뿐, 사용자의 의도를 충족시키지 못할 수 있습니다.

2. 🧑‍🏫 선호 학습 (Preference Learning)

선호 학습은 **인간의 평가(피드백)**를 기반으로 모델이 생성한 응답의 품질을 평가하는 **보상 모델 (Reward Model, $\text{RM}$)**을 훈련시키는 과정입니다.

구분 |내용
|--|--|
목표 | 여러 개의 가능한 응답 중에서 인간이 선호하는 응답을 구별하고 점수를 매기는 보상 함수를 만듭니다.
방법 | 같은 프롬프트에 대해 모델이 생성한 여러 응답을 인간 평가자에게 제시하고, "응답 $\text{A}$가 응답 $\text{B}$보다 좋다"는 식의 비교 데이터를 수집하여 $\text{RM}$을 훈련시킵니다.
결과 | 응답에 점수(스칼라 값)를 부여하여, 모델이 강화 학습 단계에서 학습 목표로 삼을 수 있는 $\text{RM}$이 탄생합니다.

⭕ **선호 학습의 역할**

선호 학습을 통해 생성된 보상 모델은 정량화하기 어려운 '유용성', '자연스러움', '안전성' 같은 추상적인 개념을 수학적 점수로 변환하여 $\text{LLM}$을 정렬(Alignment)시키는 핵심 도구로 사용됩니다.

3. 🧑‍🏫 정렬 학습 (Alignment Learning)

정렬 학습은 가장 포괄적인 개념으로, $\text{LLM}$이 인간의 의도, 가치, 윤리 기준에 부합하도록 만드는 전체 프로세스를 의미합니다.

구분 | 내용
|--|--|
목표 | 모델을 HHH 원칙($\text{Helpful, Harmless, Honest}$)에 맞춥니다. 즉, 도움이 되고, 해롭지 않으며, 진실한 응답을 하도록 만듭니다.
방법 | 가장 대표적인 방법은 **RLHF (Reinforcement Learning from Human Feedback, 인간 피드백 기반 강화 학습)**입니다. $\text{RLHF}$는 다음 단계를 포함합니다.
$\text{RLHF}$ 구성 요소 |  1. 사전 학습 (Pre-training)으로 기본 언어 능력 확보. 2. 지시 학습 (Instruction Tuning)으로 명령 수행 능력 강화. 3. 선호 학습 (Preference Learning)으로 보상 모델 ($\text{RM}$) 구축. 4. 강화 학습 ($\text{RL}$)을 통해 $\text{RM}$의 점수가 높아지는 방향으로 모델을 미세 조정.

⭕ **정렬 학습의 역할**

정렬 학습은 $\text{LLM}$이 고도화될수록 더욱 중요해집니다. 이는 모델의 창의성과 지능을 유지하면서도, 유해하거나 편향되거나 위험한 내용을 생성하는 것을 방지하는 최종 안전장치 역할을 합니다.

📝 요약 비교표

방법론 | 주된 목표 | 학습에 사용하는 데이터 | 최종 결과
|--|--|--|--|
지시 학습 | 특정 지시를 이해하고 따르는 능력 부여 | (지시, 응답) 쌍의 데이터셋 | 지시를 잘 따르는 파인 튜닝된 $\text{LLM}$
선호 학습 | 인간이 선호하는 응답의 품질을 평가 | (프롬프트, 응답 A, 응답 B) 비교 데이터 | 응답의 품질을 점수 매기는 보상 모델 ($\text{RM}$)
정렬 학습 | $\text{LLM}$을 인간의 가치에 부합하도록 조정 | 지시 학습, 선호 학습 및 강화 학습 과정 전체 | 유용성, 안전성, 윤리성을 갖춘 최종 $\text{LLM}$

# ⭐ 거대 언어 모델 (LLM) 디코딩 알고리즘 정리

### 디코딩 알고리즘은 $\text{LLM}$이 다음 토큰에 대한 확률 분포를 출력했을 때, 최종적으로 어떤 토큰을 선택하여 문장을 완성할지 결정하는 방법을 정의합니다.

1. 🧑‍🏫 그리디 탐색 (Greedy Search)

가장 단순하고 빠르게 동작하는 방법입니다.

구분 | 내용
|--|--|
원리 | 매 스텝마다 가장 높은 확률을 가진 토큰을 무조건적으로 선택합니다.
특징 | 생성 속도가 가장 빠릅니다.
단점 | 지역 최적해(Local Optimum)에 빠지기 쉬워, 전체 문장 단위에서는 가장 최적의 결과를 보장하지 못합니다. 반복적인 구문을 생성하거나 품질이 떨어지는 결과를 만들 수 있습니다.

2. 🧑‍🏫 빔 탐색 (Beam Search)

그리디 탐색의 단점을 보완하기 위해 여러 개의 후보 시퀀스를 동시에 탐색하는 방법입니다.

구분 | 내용
|--|--|
원리 | 각 스텝에서 확률이 높은 $k$개의 후보 경로(Beam)를 유지하고, 다음 스텝에서 이 $k$개 경로 각각에서 생성될 수 있는 모든 경우를 탐색한 후, 다시 가장 확률 높은 $k$개의 경로만 선택하여 다음 스텝으로 넘어갑니다.
특징 | 그리디 탐색보다 더 높은 품질의, 전체적으로 일관성 있는 문장을 생성할 가능성이 높습니다.
단점 | $k$값이 커질수록 탐색 시간이 오래 걸립니다. 또한, 확률이 높은 시퀀스에만 집중하므로 결과의 다양성이 낮습니다.
사용 | 주로 번역, 요약 등 품질과 정확성이 중요한 작업에 사용됩니다.

3. 🧑‍🏫 샘플링 (Sampling)

결정론적으로 토큰을 선택하는 그리디/빔 탐색과 달리, 확률 분포를 기반으로 무작위성을 도입하여 결과를 생성하는 방법입니다.

3-1. 기본 샘플링 (Basic Sampling)

구분 | 내용
|--|--|
원리 | 확률 분포를 그대로 사용하여 확률에 비례하게 토큰을 무작위로 추출합니다.
특징 | 결과의 다양성이 매우 높습니다. 창의적인 텍스트 생성에 유리합니다.
Temperature | 샘플링 시 무작위성의 정도를 조절하는 파라미터입니다. $T=1$은 기본 분포를 사용하고, $T<1$은 분포를 뾰족하게(확률 높은 토큰 선호), $T>1$은 분포를 평평하게(무작위성 증가) 만듭니다.

3-2. Top-K 샘플링 (Top-K Sampling)

기본 샘플링에서 발생할 수 있는 품질 저하를 막기 위해 상위 $K$개의 후보만 고려합니다.

구분 | 내용
|--|--|
원리 | 전체 어휘 중에서 가장 높은 확률을 가진 $K$개의 토큰만을 선택하여 이들 사이에서만 샘플링합니다. 나머지 토큰들의 확률은 0이 됩니다.
특징 | 희귀하거나 이상한 단어가 선택되는 것을 막아 품질을 유지하면서도 어느 정도의 다양성을 확보할 수 있습니다.

3-3. Top-P 샘플링 / 핵 샘플링 (Top-P Sampling / Nucleus Sampling)

Top-K의 고정된 $K$ 값 문제를 해결하고, 문맥에 따라 동적으로 후보군 크기를 조절합니다.

구분 | 내용
|--|--|
원리 | 확률이 높은 토큰부터 누적합을 계산하여, 누적 확률이 $P$에 도달하는 최소한의 토큰 집합($P \in [0, 1]$)만 선택하여 샘플링합니다.
특징 | 특정 단어의 확률이 매우 높을 때는 적은 수의 토큰만 고려하고, 확률 분포가 평평할 때는 많은 수의 토큰을 고려하여 문맥 적응성이 뛰어납니다. 가장 널리 사용되는 샘플링 방법 중 하나입니다.

💡 요약 비교: 성능 지표

알고리즘 | 속도 | 품질/정확성 | 다양성/창의성 | 적합한 작업
|--|--|--|--|--|
그리디 탐색 | 가장 빠름 | 낮음 (반복적) | 가장 낮음 | 짧고 단순한 응답
빔 탐색 | 느림 | 높음 (전역 최적화) | 낮음 | 번역, 코드 생성 (결정론적 작업)
Top-K / Top-P 샘플링 | 빠름 | 높음 (제어 가능) | 높음 | 대화, 스토리텔링, 창의적 글쓰기
