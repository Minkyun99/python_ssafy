### 원-핫 인코딩
1. 원-핫 인코딩이란?
- 규칙 기반 혹은 통계적 자연어처리 연구의 대다수는 단어를 원자적(쪼갤 수 없는) 기호로 취급함
    → 예 : hotel, conference, walk 같은 단어들
- 벡터 공간 관점에서 보면, 이는 한 원소만 1이고 나머지는 모두 0인 벡터를 의미함
- 차원 수(=단어 사전 크기)는 대략 다음과 같음
    → 음성 데이터(2만) - Penn Treebank(PTB) 코퍼스(5만) - big vocab(50만) - Google 1T(1300만)
- 이를 **원-핫 표현**이라고 부르고, 단어를 **원-핫 표현으로 바꾸는 과정을 원-핫 인코딩**임

2. 원-핫 인코딩의 문제점
- 예시 : 웹검색
    → [삼성 노트북 배터리 사이즈] == [삼성 노트북 배터리 용량]
    → [갤럭시 핸드폰] == [갤럭시 스마트폰]  
    ![image.png](./img/ai58.png)  
- 검색 쿼리 벡터와 대상이 되는 문서 벡터들이 서로 직교하게 되어, 원-핫 벡터로는 유사도를 측정할 수 없음
- 원-핫 인코딩과 같은 전통적인 텍스트 표현 방식에는 여러 한계가 존재함
    → 차원의 저주(Curse of Kimensionality)  
        - 고차원의 희소 벡터를 다루기 위해선 많은 메모리가 필요함  
        - 차원이 커질수록 데이터가 점점 더 희소해져 활용이 어려움  
    → 의미적 정보 부족  
        - 비슷한 단어라도 유사한 벡터로 표현되지 않음  
        - 예시 : '은행'과 '금융'은 의미적으로 밀접하지만, 원-핫 인코딩에서는 전혀 무관한 벡터로 취급됨  

### 워드 임베딩
1. 주변 단어들을 활용해보기
- 단어를 주변 단어들로 표현하면, 많은 의미를 담을 수 있음
    → 'You shall know a word by the company it keeps'
- 현대 통계적 자연어 처리에서의 가장 성공적인 아이디어 중 하나임
    → **정부 가계 부채 문제를 해결하기 위해 은행 대출 규제를 강화함**
    → **인공지능 기반 챗봇이 은행 상담 서비스에 도입됨**
    → 이런 주변 단어들로 은행의 의미를 나타낼 수 있음
- 단어를 단어들 사이의 의미적 관계를 포착할 수 있는 **밀집(dense)되고 연속적/분산적(distributed) 벡터 표현으로 나타내는 방법**임
    → 원-핫 인코딩에선 은행과 금융이 완전히 독립적인(무관한) 벡터로 표현되었지만,
    → 워드 임베딩에선 두 단어의 벡터가 공간상 서로 가깝게 위치하며, 이를 통해 의미적 유사성을 반영할 수 있음

2. 대표적인 워드 임베딩 기법 - Word2Vec
- Word2Vec은 2013년 Google에서 개발한 워드 임베딩 기법임
- 단어의 표현을 간단한 인공 신경망을 이용해서 학습함  

3. Word2Vec의 아이디어
- Word2Vec의 아이디어는 각 단어와 그 주변 단어들 간의 관계를 예측한다는 것임
- Word2Vec엔 두 가지 알고리즘이 존재함
    → Skip-grams(SG) 방식
        - 중심 단어를 통해 주변 단어들을 예측하는 방법임  
        - 단어의 위치(앞/뒤)에 크게 구애 받지 않음
    → Continuous Bag of Words(CBOW) 방식
        - 주변 단어들을 통해 중심 단어를 예측하는 방법임
        - 문맥 단어들의 집합으로 중심 단어를 맞춤  

4. Skip-grams(SG) : 중심 단어를 통해 주변 단어 예측하기

# ⭐ 순차적 데이터

1. 순차적 데이터란 무엇일까?
- 자연엔 수많은 순차적 데이터(Sequential Data)가 존재함
- 순차적 데이터란 데이터가 입력되는 순서와 이 순서를 통해 입력되는 데이터들 사이의 관계가 중요한 데이터임
- 예시  
![image.png](./img/ai64.png)  

2. 순차적 데이터의 특징
- 순차적 데이터는 다른 데이터와 구별되는 특징들이 존재함
- 특징
    1. 순서가 중요함
        - 데이터의 순서가 바뀌면 의미가 달라짐
        - 예시 : 나는 너를 사랑해 != 너는 나를 사랑해
    2. 장기 의존성(Long-term dependency)
        - 멀리 떨어진 과거의 정보가 현재/미래에 영향을 줌
        - 예시 : '여러 개의 문 중 파란 문을 열고 안으로 들어가면, 너는 (?)를 찾게 될거야'
    3. 가변 길이(Variable length)
        - 순차 데이터는 길이가 일정하지 않고, 단어 수도 제각각임

3. 순차적 데이터를 처리하려면?
- 따라서, 순차적 데이터를 처리하려면 일반적인 모델들(예 : 선형회귀, MLP 등)로는 불가능함
- Sequential Models 필요함
    → 예시 : RNN, LSTM, Transformer 등등

### 🧑‍🏫 RNN
1. 전통적인 인공신경망
- 전통적인 인공신경망(MLP, CNN)들은 고정된 길이의 입력을 받아 **가변 길이의 데이터**를 처리하기에 적합하지 않음

2. RNN이란?
- 하지만, RNN은 가변 길이의 입력을 받을 수 있고, 이전 입력을 **기억**할 수 있기 때문에, **순차적 데이터 처리**에 적합한 아키텍처임  
![image.png](./img/ai67.png)  
![image.png](./img/ai68.png)  

3. RNN 아키텍처 설명
- 전통적인 신경망(MLP, CNN 등)과 달리, RNN은 이전 시점의 정보를 담는 hidden state를 가지고 있음
- 따라서, 입력 시퀀스 벡터 x를 처리할 때, 각 시점마다 recurrence 수식을 적용하여 hidden state를 업데이트함    
![image.png](./img/ai69.png)    
![image-2.png](./img/ai70.png)  

4. RNN의 특징
- RNN은 한 번에 하나의 요소를 처리하고, 정보를 앞으로 전달함
- 펼쳐서 보면, RNN은 각 층이 하나의 시점을 나타내는 깊은 신경망처럼 보임
- RNN은 hidden state를 유지하면서 가변 길이 데이터를 처리할 수 있음
- RNN의 출력은 과거 입력에 영향을 받는다는 점에서, feedforward 신경망과 다름

5. RNN의 한계 : 기울기 소실(vanishing gradient) 문제
- 기울기 소실 문제란?
    - 딥러닝에서 연전파 시 앞쪽 층의 기울기가 0에 가까워져서 장기 의존성 학습이 어려워지는 현상
- 왜 일어날까?
    1. 역전파 과정에서 작은 값들이 계속 곱해짐
    2. 과거 시점에서 온 오차 신호는 갈수록 더 작은 기울기를 갖게 됨
    3. 결국 파라미터들이 장기 의존성은 학습하지 못하고, 단기 의존성만 포착하게 됨

### LSTM
1. LSTMs란?
- 기울기 소실 문제를 해결하기 위해, 1997년에 제안된 RNN의 한 종류
- LSTMs의 특징
    - 시점 t에서 RNN은 길이가 n인 벡터 hidden state ht와 cell state Ct를 가짐
        - Hidden state는 short-term information을 저장함
        - Cell states는 long-term information을 저장함
    -LSTMs는 cell state에서 정보를 읽고, 지우고, 기록할 수 있음

2. RNNs VS LSTMs

3. LSTMs
- 3가지 게이트를 통해 어떤 정보를 지우고, 쓰고, 읽을지 결정함
    → Forget gate : 이전 cell state에서 무엇을 버리고 무엇을 유지할지 결정함  
    → Input gate : 새 정보 중 얼마나 cell state에 쓸지 결정
    → Output gate : cell state 중 얼마나 hidden state로 내보낼지 결정

2. 대표적인 언어모델 -  N-gram 언어모델
- n-gram이란, 연속된 n개의 단어 묶음을 말함
    → unigrams : 'The', 'Students', 'opened', 'opened their'  
    → bigrams : 'The students', 'students opened', 'opened their'  
    → trigrams : 'The students opened', 'students opened their'  
    → four-grams : 'The students opened thier'  
- 다양한 n-gram이 얼마나 자주 등장하는지 통계를 수집하고, 이를 활용해 다음 단어를 예측함
- 예시 : 4-gram 언어모델

- 말뭉치 정보

    → 'students opened their books' - 400번 등장  
        - P(books | students opened their) = 0.4
    → 'students opened their exams' - 100번 등장
        - P(exams | students opened their) = 0.1

- 이런 한계로 인해, 이후의 기계 번역 연구는 Neural Machine Translation(NMT)로 넘어가게 됨

### 🧑‍🏫 Seq2Seq
1. Neural Machine Translation이란?
- Neural Machine Translation이란 인공 신경망을 이용해 기계 번역을 수행하는 방법임
- 이때 사용되는 신경망 구조를 sequence-to-sequence(Seq2Seq)라 하며, 두 개의 RNNs으로 이루어짐
    → 2014년 Google의 'Sequence to Sequence Learning with Neural Networks'라는 논문에서 처음 소개됨

2. Translation이 어려운 이유
- 번역 문제는 입력과 출력의 길이가 다를 수 있음  
    → 영어 : the black cat drank milk(5개의 단어)   
    → 프랑스어 : le chat noir a bu du lait(7개의 단어)  
    **→ 따라서, NMT에서는 길이가 다른 시퀀스 간의 매핑을 처리할 수 있어야 함**  

3. Seq2Seq의 아이디어
- 2개의 LSTM을 이용하자
    → 한 LSTM은 입력 시퀀스를 한 타임스텝씩 읽어 고정된 차원의 큰 벡터 표현을 얻기 (Encoder)  
    → 다른 LSTM은 앞에서 얻은 벡터로부터 출력 시퀀스를 생성하기 (Decoder)  

4. Seq2Seq Architecture
- Seq2Seq는 Encoder와 Decoder로 이루어짐
    → Encoder는 입력 문장에 담긴 정보를 인코딩함  
    → Decoder는 인코딩된 정보를 조건으로 하여 타겟 문장(출력)을 생성함    
    ![image.png](./img/ai84.png)  

5. Seq2Seq의 다양한 적용
- Seq2Seq 구조는 기계번역 외에도 다양한 태스크에 적용할 수 있음
    → 📝 요약 : 긴 길이의 문서를 읽고, 짧은 길이의 문장으로 요약된 텍스트를 출력하는 태스크
    → 🗨️ 대화 : 사용자의 발화를 기반으로, 맥락에 맞는 대답(출력 텍스트)을 생성하는 태스크
    → 💻 코드 생성 : 자연어로 작성된 설명 혹은 명령어를 입력 받아, 그에 대응하는 프로그래밍 코드 혹은 쿼리를 출력하는 태스크

6. Seq2Seq 학습 수행(Teacher Forcing)
- Seq2Seq 모델은 인코더와 디코더가 하나의 통합 네트워크로 연결되어 있음
- 디코드에서 발생한 오차는 역전파 과정을 통해 입력을 처리한 이코더까지 전달되어 전체 네트워크가 End-to-End로 동시에 최적화됨  
![image.png](./img/ai85.png)  
- 학습 초반에는 모델의 예측 능력이 떨어지기 때문에 학습이 불안정할 수 있음
- Teacher Forcing이란?
    → 모델이 스스로 예측한 단어 대신 정답 단어를 디코더 입력으로 강제로 넣어줌으로써 훨씬 안정적이고 빠르게 학습을 수행하는 방법임
- 토큰을 출력하는 방법 중 하나로, 각 단계에서 가장 확률이 높은 단어를 선택함
- 한계
    → 되돌리기가 불가능함
    → 예시 : le______ → le chien _________ → 
    → 위의 출력이 오답이더라도 이전으로 되돌아갈 방법이 없음  
    ![image.png](./img/ai86.png)  
- Beam Search
    1. 매 단계마다 k개의 가장 유망한 후보 유지
    2. 후보가 <EOS>에 도달하면, 완성된 문장으로 리스트 추가
    3. <EOS> 문장이 충분히 모이면 탐색 종료
    4. 각 후보들의 점수를 로그 확률의 합으로 구해 최종 선택

### 🧑‍🏫 Attention
1. Seq2Seq의 한계 : the bottleneck problem
- Botteneck problem이란?  
    → 인코더는 입력 문장 전체를 하나의 벡터로 요약하는데, 마지막 hidden state에 문장의 모든 의미 정보가 담김  
    → 고정 길이 벡터 하나에 모든 문장의 의미를 압축하다 보니 정보 손실이 생길 수 잇는데, 이를 bottleneck   problem이라고 함  
    ![image.png](./img/ai87.png)    
 
2. Attention의 인사이트
- Attention은 디코더가 단어를 생성할 때, 인코더 전체 hidden state 중 필요한 부분을 직접 참조할 수 있도록 함
- 즉, 매 타임스텝마다 '어떤 단어/구절에 집중할지'를 가중치롤 계산해, bottelneck 문제를 완화함  
![image.png](./img/ai88.png)  

3. Attention의 효과
- Attention mechanism은 많은 장점이 존재함
    1. NMT 성능 향상
        - 디코더가 소스 문장 전체가 아닌, 필요한 부분에만 집중할 수 있음
    2. Bottleneck Problem 해결
        - 디코더가 인코더의 모든 hidden states에 직접 접근할 수 있음
    3. Vanishing Gradient Problem 완화
        - Attention은 멀리 떨어진 단어도 직접 연결할 수 있게 해줌  
        ![image.png](./img/ai89.png)  
        ![image.png](./img/ai90.png)  

4. Attention의 효과 : 해석 가능성(Interpretability)
- Attention 분포를 보면, decoder가 어떤 단어를 생성할 때, 입력 문장의 어느 부분에 집중했는지 확인할 수 있음
- 즉, 모델이 내부적으로 참고한 근거를 사람이 파악할 수 있음
    → 모델의 의사결정 과정을 해석할 수 있는 단서  

5. Attention의 효과 : 정렬(Alignment)
- 기계번역에서는 전통적으로 단어 alignment 모델을 따로 학습해야 함
- 하지만, attention을 통해 decoder가 필요한 입력 단어에 자동으로 집중하기 때문에, 단어와 단어 간의 매핑 관계를 자연스럽게 학습함

6. Attention : Query와 Values
- Seq2Seq에서 attention을 사용할 때, 각 decoder의 hidden state와 모든 encoder의 hidden states 간의 관계를 Query와 Vaules의 관계로 볼 수 있음
- 이 관점에서 Attention 과정을 정리해보면 :
    1. Query와 Values 사이 유사도 점수(score) 계산 (예 : dot-product, multiplication, additive 등)
    2. softmax를 통해 확률 분포(attention distribution) 얻기
    3. 분포를 이용해 values를 가중합 → context vector(attention output)  
    ![image.png](./img/ai91.png)  


# ⭐ Transformer

### 🧑‍🏫 self-Attention
1. RNN이 꼭 필요할까?
- RNN은 정보를 hidden state로 전달하며 순차적 의존성을 형성함
- 그에 반해 Attention은 필요한 순간, 입력 전체에서 직접 정보를 전달함
→ RNN이 하던 '정보 전달'을 Attention이 더 효율적으로 수행할 수 있다면, 굳이 recurrence가 필요할까?  
![image.png](./img/ai92.png)  

2. RNN의 한계점1 - 장기 의존성
- RNN은 왼쪽에서 오른쪽으로 순차 전개되어, 먼 단어 쌍이 상호작용하려면 시퀀스 길이 만큼의 단계를 거쳐야 함
- 따라서, 길어진 단계만큼 기울기 소실 혹은 폭발 문제가 발생해 장기 의존성을 학습하기 어려움
- 또한 RNN은 입력된 선형 순서를 강하게 반영하기 때문에, 언어의 비선형적 의존성을 잘 잡아내지 못함
    → 예시 : 그 책은 오랫동안 방치되어 먼지가 많이 쌓인 탓에 ... 매우 더러웠다.  
    ![image.png](./img/ai93.png)  

3. RNN의 한계점2 - 병렬화
- Forward와 Backward pass 모두 시퀀스 길이 만큼의 단계가 필요함
- 이처럼, 순차적인 연산이 진행되기 때문에, 병렬화가 불가능함
- 이는, 병렬 연산에 강한 GPU 활용을 어렵게 만들며, 대규모 데이터 학습에 비효율적임  
![image.png](./img/ai94.png)  

4. 그렇다면 Attention은 어떨까?
- Attention은 각 단어의 표현을 query로 두고, value 집합으로부터 필요한 정보를 직접 불러와 결합함
- 이러한 Attention 메커니즘을 encoder-decoder 간이 아닌, 한 문장 내부에서 적용한다면 어떨까?
    → Self-Attention

5. Self-Attention의 장점은?
- 순차적으로 처리해야 하는 연산 수가 시퀀스 길이에 따라 증가하지 않음
- 최대 상호작용 거리 = O(1) → 모든 단어가 각 층에서 직접 상호작용함

8. Self-Attention - Key, Query, Value
- Attention에서 각 단어 'i'를 표현하는 Query와 Value 벡터가 있었듯이, Self-Attention에서는 각 단어 'i'를 표현 하는 Query, Key, Value 벡터가 존재함  
    ![image.png](./img/ai103.png)  
    1. 각 단어를 Query, Key, Value 벡터로 변환함
        - Query 벡터
            → 단어 i가 다른 단어로부터 어떤 정보를 찾을지를 정의하는 벡터    
        - Key 벡터
            → 단어 i가 자신이 가진 정보의 특성을 표현하는 벡터  
        - Value 벡터
            → 실제로 참조되는 정보 내용을 담고 있는 벡터  
        - Attention에서 각 단어 'i'를 표현하는 Query와 Value 벡터가 있었듯이, Self-Attention에서는 각 단어 'i'를 표현하는 Query, Key, Value 벡터가 존재함
    2. Query, Keys 간의 유사도를 계산해, softmax로 확률분포를 구함  
    ![image.png](./img/ai104.png)  
    3. 각 단어의 출력을 Values의 가중합으로 계산함  
    ![image.png](./img/ai105.png)  

9. Self-Attention의 한계
- 이렇듯, Self-Attention은 단어 간 관계를 효율적으로 잡아내는 강력한 메커니즘이지만, 한계가 존재함
- 한계
    1. 순서 정보 부재 : 단어 간 유사도만 계산하기 때문에, 단어의 순서를 고려하지 않음
    2. 비선형성 부족 : Attention 계산은 본질적으로 가중 평균 연산이라는 선형 결합에 불과하기 때문에, 복잡한 패턴이나 깊은 표현력을 담기 어려움
    3. 미래 참조 문제 : 언어 모델은 시퀀스를 왼쪽에서 오른쪽으로 생성해야 하지만, Self-Attention은 모든 단어를 동시에 보기 때문에, 아직 생성되지 않아야 할 미래 단어를 참조함

10. Self-Attention의 한계 해결1 - Positional Encoding
- 순서 정보 부재 문제를 해결하기 위해, Positional Encoding이라는 기법을 사용함
- 각 단어 위치 i를 나타내는 위치 벡터를 정의해, 단어 임베딩 값에 더해 최종 입력으로 사용함
- 두 가지 방법이 존재함
    1. Sinusoidal Position Encoding
        - 서로 다른 주기의 사인/코사인 함수를 합성해 위치 벡터를 만드는 방법
    2. Learned Absolute Position Embedding
        -  위치 벡터를 모두 학습 파라미터로 설정해 학습 과정에서 데이터에 맞춰 최적화하는 방법
    
11. Self-Attention의 한계 해결2 - Feed-Forward Network 추가하기
- Self-Attention 연산은 비선형 변환이 없어, 복잡한 패턴 학습에 한계가 존재함
- 따라서, 각 단어 출력 벡터에 Feed-Forward Network(Fully Connected + ReLU 등)을 추가해 Self-Attention이 만든 표현을 깊고 비선형적인 표현으로 확장함  
![image.png](./img/ai106.png)  

12. Self-Attention의 한계 해결3 - Masked Self-Attention
- 단어를 생성할 때는 한 단어씩 순차적으로 미래 단어를 예측해야 하지만, Self-Attention은 기본적으로 모든 단어(미래 포함)를 동시에 참조함
- 따라서, Attention Score를 계산할 때, 미래 단어에 해당하는 항목을 -무한대로 설정해, 계산을 수행할 때 반영되지 않도록 함  
![image.png](./img/ai107.png)  

13. Self-Attention 정리
- Self-Attention은 문장 내 모든 단어가 서로 직접 상호작용하여, **1) 장거리 의존성을 효율적으로 포착하고** **2) 병렬 처리**를 가능하게 하는 메커니즘임
- 한계 해결 방법
    1. 순서 정보 부재 → Positional Encoding 
    2. 비선형 부족 → Feed-Forward Network 추가
    3. 미래 참조 문제 → Masked Self-Attention  
![image.png](./img/ai108.png)  



### Transformer

1. Transformer
- Transformer는 encoder-decoder 구조로 설계된 신경망 모델임
    → encoder : 입력 문장을 받아 의미적 표현으로 변환을 수행함
    → decoder : 인코더의 표현과 지금까지 생성한 단어들을 입력 받아, 다음 단어를 예측함
    → **이 중 decoder가 언어 모델과 같은 방식으로 동작함**

2. Multi-Headed Attention
- 문장에서 같은 단어라도 여러 이유(문법적 관계, 의미적 맥락 등)로 다른 단어에 주목할 수 있음
- 하지만, 단일 Self-Attention Head로는 한 가지 관점에서의 단어 간 관계 밖에 파악할 수 없음
➡️ 따라서, 여러 Attention Head를 두어 다양한 관점에서 동시에 정보를 파악함
    - Attention Head 1 - 단어의 문맥적 관계에 Attention    
    - Attention Head 2 - 단어의 시제에 Attention  
    - Attention Head 3 - 명사에 Attention    
![image.png](./img/ai109.png)  

3. Scaled Dot Product
- Query와 Key의 차원이 커질수록, 두 벡터의 내적 값도 자연스럽게 커짐
- 이 값이 너무 크면 softmax 함수가 출력이 지나치게 뾰족해져, 미세한 변화에도 큰 차이가 발생하고 gradient vanishing 문제가 생길 수 있음  
![image.png](./img/ai110.png)  
- 따라서, 내적 값을 그대로 사용하지 않고, 나눠주어 스케일을 조정함
- 이렇게 하면 값이 안정적으로 분포되어 학습이 훨씬 더 빠르고 안정적으로 진행됨  
![image.png](./img/ai111.png)  

4. Residual Connection
- 깊은 신경망은 층이 깊어질수록 학습이 어려워짐 (Gradient vanishing / exploding)
- 따라서, 단순히 Layer의 출력만 사용한다면 정보가 소실되어, layer가 전체를 예측하는 것이 아니라, 기존 입력과의 차이만 학습하도록 하는 residual connection을 사용함  
![image.png](./img/ai112.png)  

5. Layer Normalization
- 층이 깊어질수록, hidden vector 값들이 커졌다 작아졌다 하면서 학습이 불안정함
- Layer Normalization은 각 레이어 단위에서 hidden vector 값을 정규화해, 안정적이고 빠른 학습을 도움

6. Decoder
- Transformer의 decoder는 여러 개의 decoder 블록들을 쌓아 올려서 만든 구조임
- 각 블록은 다음으로 구성됨
    ➡️ Masked Self-Attention(Multi-Head)    
        - 미래 단어를 보지 않도록 마스크를 씌운 Multi-Head Self-Attention    
    ➡️ Add & Norm (Residual Connection + Layer Normalization)  
    ➡️ Feed-Forward Network  
        - 각 위치 별 비선형 변환을 수행함  
    ➡️ Add & Norm (Residual Connection + Layer Normalization)  
    **➡️ 언어 모델처럼 단방향 문맥만 활용**  

7. Encoder
- 그에 반해 Transformer의 encoder는 양방향 문맥을 모두 활용할 수 있음  
    ➡️ 입력 문장을 의미적 표현으로 변환할 수 있음    
    ➡️ 각 단어가 양방향 문맥을 모두 반영한 벡터로 인코딩 됨  
- Decoder와의 차이점은 Self-Attenction에서 masking을 제거한 것 뿐임    
![image.png](./img/ai113.png)  

8. Encoder-Decoder
- 기계 번역에서 Seq2Seq 모델을 사용했던 것처럼, Transformer에서도 이해를 위한 encoder와 생성을 위한 decoder로 이뤄진 encoder-decoder 구조를 채택함
- decoder는 단순 Self-Attention만 하는 것이 아니라, encoder의 출력 표현을 참조하는 Cross-Attention을 추가하여 입/출력을 연결함

9. Cross-Attention
- Cross-Attention을 수행할 때는 Self-Attention과는 다르게 Query는 decoder에서, Key와 Value는 ecoder에서 가져옴

10. Transformer의 놀라운 결과
- Neural Machine Translation task에서 당시 최고 성능을 달성했을 뿐 아니라, 가장효율적인 학습으로 비용까지 절감할 수 있었음

11. Transformer와 사전학습(pretraining)
- Trnasformer의 등장은 대부분의 최신 모델들이 성능 향상을 위해 사전학습(pretraining)을 결합하도록 했음
- 또한, 뛰어난 병렬 처리 능력 덕분에 대규모 사전 학습에 적합하여 NLP의 표준 아키텍처로 자리 잡음

# 사전 학습 기반 언어 모델

### 사전학습이란?
1. 사전학습이란?
- 사전학습이란 대규모 데이터 셋을 이용해, 모델이 데이터의 일반적인 특징과 표현을 학습하도록 하는 과정임
- 특히 언어 모델은, 인터넷의 방대한 텍스트(웹 문서, 책, 뉴스 등)을 활용해 비지도학습 방식으로 학습되어, 일반적인 언어 패턴, 지식, 문맥 이해 능력을 습득함

2. 사전학습의 관점에서 워드 임베딩 VS 언어 모델
- 워드 임베딩의 경우 사전학습을 통해 단어의 의미를 학습하지만 한계가 존재함
    1. 다운스트림 태스크 (예 : 텍스트 분류)들에 적용하기엔 학습되는 데이터의 양이 적어, 언어의 풍부한 문맥 정보를 충분히 학습할 수 없음
    2. 연결된 네트워크가 무작위 초기화되어 학습 효율이 낮고, 많은 데이터와 시간이 필요함
- 언어 모델의 경우, 모델 전체를 사전학습을 통해 학습하기 때문에, 강력한 NLP 모델을 구축하는 데에 이점이 있음
    1. 언어에 대한 표현 학습용으로 적합함
    2. 파라미터 초기화의 관점에서 효과적임
    3. 언어에 대한 확률 분포 학습을 통해 샘플링, 생성에 사용할 수 있음

3. 언어모델의 사전학습
- 과거 단어들이 주어졌을 때, 다음 단어의 확률분포를 모델링하는 방법을 배움으로서, 사전학습을 수행할 수 있음
- 인터넷 대규모 텍스트 코프스에서 언어모델링 학습을 수행 후학습된 네트워크 파라미터를 저장해, 다양한 다운스트림 태스크에 활용함

4. 사전 학습 ➡️ 파인튜닝(fine-tuning) 패러다임

### Encoder 모델  
1. Encoder 모델
![image.png](./img/ai117.png)  

2. Encoder 모델의 사전 학습
- Encoder 모델은 양방향 문맥을 모두 활용하기 때문에, 전통적인 언어모델과는 차이점이 있음
- 따라서, Encoder 모델의 사전 학습을 위해선 입력 단어의 일부를 [MASK] 토큰으로 치환해, 모델이 이 [MASK] 자리에 올 단어를 예측하도록 학습하는 방법을 사용할 수 있음
- 이를 Masked Language Model이라 하며, 대표적인 모델이 **'BERT'**임

3. BERT
- BERT는 2018년 Google에서 공개한 transformer 기반의 모델로, Masked LM 방법으로 사전학습을 수행함

4. BERT의 학습 방법1 - Masked LM
- 학습 방식은 다음과 같음
    ➡️ 입력 토큰의 15%를 무작위로 선택함
    ➡️ [MASK] 토큰 치환(80%), 랜덤 토큰 치환(10%), 그대로 두기(10%)
- 모델이 마스크된 단어에만 집중하지 않고, 다양한 문맥 표현을 학습해 더 강건한(robust) 표현을 학습할 수 있도록 했음  
![image.png](./img/ai118.png)  
![image.png](./img/ai119.png)  

5. BERT의 학습 방법2 - Next Sentence Prediction (NSP)
- 또한, BERT는 입력을 두 개의 연속된 텍스트로 받아, 두번째 문장이 첫번째 문장의 실제 다음 문장인지 여부를 예측하는 Next Sentence Prediction(NSP)을 수행했음  
![image.png](./img/ai120.png)  
- NSP를 통해 문장 간 관계를 학습하여 문맥적 추론 및 문장 수준 이해 태스크에 도움이 되도록 설계됨
- 예시 :
    1. 자연어 추론
    2. Parapharse detection
    3. 질의응답 (Question Answering)  
![image.png](./img/ai121.png)  

6. BERT의 다운스트림 태스크
- BERT는 이렇게 MLM과 NSP 두 가지 태스크를 통해 학습함
- [CLS] 토큰은 NSP 태스크 용으로 학습되며, 다른 토큰들은 MLM 태스크 용으로 학습됨  
![image.png](./img/ai122.png)  

7. BERT의 다운스트림 태스크 - Sentence Level
- 두 문장 관계 분류 태스크
    1. MNLI
        - 전제(Premise) : 여러 남자들이 뛰고 있는 축구 경기
        - 가설(Hypothesis) : 몇몇 남자들이 스포츠를 하고 있음
        - 분류 : {함의(Entailment), 모순(Contradiction), 중립(Neutral)}
    2. QQP
        - Q1 : 주식 투자 방법은 어디서 배울 수 있나요?
        - Q2 : 주식에 대해 더 배우려면 어떻게 해야 하나요?
        - 분류 : {중복(Duplicate), 비중복(Not Duplicate)}
- 단일 문장 분류 태스크
    1. SST2
        - 문장 : 이 영화에는 재미있는 요소들이 풍부함
        - 분류 : {긍정(Positive), 부정(Negative)}
- QA태스크:
    1. SQuAD
        - 질문 : 뉴욕 자이언츠와 뉴욕 제츠는 뉴욕시의 어느 경기장에서 경기를 하나요?
        - 문맥 : ... 두 팀은 뉴저지 이스트 러더퍼드에 있는 Metilife Stadium에서 홈경기를 한다...
        - 정답 : Metlife Stadium

8. BERT의 다운스트림 태스크 - Token Level
- 개체명 인식(Named Entity Recognition, NER)
    1. CoNLL 2003 NER
        - 문장 : John Smith lives in New York 
        - 라벨 :
            - John ➡️ B-PER (사람 이름의 첫 단어)
            - Smith ➡️ I-PER (사람 이름에 속하는 나머지 단어)
            - live ➡️ O (개체명이 아닌 단어)
            - In ➡️ O (개체명이 아닌 단어)
            - New ➡️ B-LOC (장소 이름의 첫 단어)
            - York ➡️ I-LOC (장소 이름에 속하는 나머지 단어)

10. BERT의 한계
- 인코더 기반 모델인 BERT는 주어진 입력을 잘 이해하도록 학습되지만, 시퀀스를 생성해야 하는 태스크에는 적합하지 않음 (예시 : 기계번역, 텍스트 생성 등)
- 생성 태스크에서는 autoregressive하게, 즉 한 번에 한 단어씩 생성해야 하는데, 이를 자연스럽게 수행하지 못하기 때문에, 생성 태스크엔 디코더 기반 모델을 주로 사용함

### Decoder 모델
최근에는 Encoder가 아닌 Decoder만 사용하여 언어모델을 개발하고 있음
- Decoder의 성능이 Encoder보다 좋기 때문에, 같은 환경이라면 Encoder를 포함시키는 것보다 해당 자원에 Decoder를 더 확장하는 것이 모델의 성능을 높일 수 있기 때문

1. Finetuning Decoder
- Transformer의 Decoder는 사전학습 단계에서 다음 단어 예측(Next Token Prediction)을 학습함
- 생성 태스크에 활용할 때 : 
    1. 사전학습 때와 동일하게 다음 단어 예측 방식을 fine-tuning함
    2. 따라서, decoder는 대화나 요약 태스크 등 출력이 시퀀스인 태스크에 자연스럽게 적합함
- 분류 태스크에 활용할 때 :
    1. 마지막 hidden state 위에 새로운 linear layer를 연결해 classifier로 사용함
    2. 이때, linear layer는 아예 처음부터 다시 학습해야 하며, fine-tuning 시 gradient가 decoder 전체로 전파됨

### In-Context Learning
1. GPT-3
- GPT-3는 2020년 OpenAI에서 공개한 모델로, GPT-2에서 모델의 parameter size를 키워(1750억개), 별도의 파인튜닝 없이 컨텍스트 안의 예시만 보고도 새로운 태스크를 수행할 수 있게 됨
- 이런 능력을 In-Context Learning이라고 함
