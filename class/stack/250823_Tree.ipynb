## ⭐ 이진 탐색 트리(BST)

##### 탐색 작업을 효율적으로 하기 위한 자료 구조
![image.png](./img/tree10.png)  
- 모든 원소는 서로 다른 유일한 키를 갖는다.
- key(왼쪽 서브트리) < key(루트 노드) < key(오른쪽 서브트리)
- 왼쪽 서브트리와 오른쪽 서브트리도 이진 탐색 트리다.
- 중위 순회하면 오름차순으로 정렬된 값을 얻을 수 있다.

##### 탐색 연산
- 루트에서 시작한다.
- 탐색할 키 값 x를 루트 노드의 키 값과 비교한다.
- (키 값 x = 루트노드의 키 값)인 경우 : 원하는 원소를 찾았으므로 탐색연산 성공
- (키 값 x < 루트노드의 키 값)인 경우 : 루트 노드의 온쪽 서브트리에 대해서 탐색연산 수행
- (키 값 x > 루트노드의 키 값)인 경우 : 루트노드의 오른쪽 서브트리에 대해서 탐색연산 수행
- 서브트리에 대해서 순환적으로 탐색 연산을 반복한다.!
![image.png](./img/tree11.png)     

##### 삽입연산
1. 먼저 탐색 연산을 수행
   - 삽입할 원소와 같은 원소가 트리에 있으면 삽입할 수 없으므로, 같은 원소가 트리에 있는지 탐색하여 확인한다.
   - 탐색에서 탐색 실패가 결정되는 위치가 삽입 위치가 된다.
2. 탐색 실패한 위치에 원소를 삽입
   - 다음 예는 5를 삽입하는 예이다. 
  
##### 이진 탐색 트리의 성능
1. 탐색, 삽입, 삭제 시간은 트리의 높이만큼 시간이 걸림
   - O(h), h : BST의 깊이
2. 평균의 경우
   - 이진 트리가 균형적으로 생성되어 있는 경우
   - O(log n)
3. 최악의 경우
   - 한쪽으로 치우친 경사 이진트리의 경우
   - O(n)
   - 순차탐색과 시간 복잡도가 같음 

## ⭐ 힙
완전 이진 트리에 있는 노드 중에서 키 값이 가장 큰 노드나 키 값이 가장 작은 노드를 찾기 위해서 만든 자료 구조  

##### 최대 힙 (max heap)
1. 키 값이 가장 큰 노드를 찾기 위한 완전 이진 트리
2. 부모 노드의 키값 > 자식 노드의 키값
3. 루트 노드 : 키 값이 가장 큰 노드

##### 최소 힙 (min heap)
1. 키 값이 가장 작은 노드를 찾기 위한 완전 이진 트리
2. 부모 노드의 키값 < 자식 노드의 키값
3. 루트 노드 : 키값이 가장 작은 노드

![image.png](./img/tree12.png)  

```python
heap = [0] * 100
last = 0

def enq(n):
    global last
    last += 1
    heap[last] = n

    c = last
    p = c // 2
    while p and heap[p] < heap[c]:
        heap[p], heap[c] = heap[c], heap[p]
        c = p
        p = c // 2

enq(2)
enq(5)
enq(7)
enq(3)
enq(4)
enq(6)
print(heap)
```

```python
# 기존 힙의 상태를 가정합니다.
# 이전에 enq 함수를 통해 heap이 다음과 같이 채워져 있다고 가정합니다.
heap = [0, 7, 6, 5, 2, 3, 4, 0, 0, 0, 0, 0, 0]
last = 6

def deq():
    global last
    if last == 0:
        print("힙이 비어있습니다.")
        return None

    # 1. 루트(가장 큰 값)를 임시 변수에 저장
    tmp = heap[1]

    # 2. 마지막 노드를 루트로 옮김
    heap[1] = heap[last]
    heap[last] = 0  # 마지막 노드 위치를 비움
    last -= 1       # 힙의 크기 감소

    # 3. 새로운 루트 노드를 제자리로 이동 (하향식 힙화)
    p = 1
    c = p * 2  # 왼쪽 자식 노드
    
    # 자식 노드가 힙의 범위를 벗어나지 않는 동안 반복
    while c <= last:
        # 오른쪽 자식이 존재하고, 왼쪽 자식보다 크면
        if c + 1 <= last and heap[c] < heap[c + 1]:
            c += 1  # 더 큰 자식으로 이동

        # 부모 노드의 값이 더 큰 자식보다 작으면 위치 교환
        if heap[p] < heap[c]:
            heap[p], heap[c] = heap[c], heap[p]
            # 위치를 교환했으므로, 다음 레벨로 이동
            p = c
            c = p * 2
        else:
            # 부모가 자식보다 크면 힙 속성이 만족되므로 반복문 종료
            break

    return tmp

# deq 함수 실행 예시
print("현재 힙:", heap[1:last+1])

removed_value = deq()
print(f"제거된 값: {removed_value}")
print("deq 후 힙:", heap[1:last+1])
print("deq 후 힙 크기:", last)

print("\n")
removed_value = deq()
print(f"제거된 값: {removed_value}")
print("deq 후 힙:", heap[1:last+1])
print("deq 후 힙 크기:", last)
```
