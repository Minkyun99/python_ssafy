```python
# 10진수를 2진수로 변환

def decimal_to_binary(n):
    binary_number = ""

    while n > 0:
        remain = n % 2
        binary_number = str(remain) + binary_number
        n = n//2

    return binary_number

print(decimal_to_binary(3))
```

```python
# 10진수를 16진수로 변환

def decimal_to_hexadecimal(n):
    hex_digit = "0123456789ABCDEF"
    hexadecimal_number = ""

    if n == 0:
        return 0
    
    while n > 0:
        remain = n % 16
        hexadecimal_number = hex_digit[remain] + hexadecimal_number
        n = n // 16
    return hexadecimal_number

print(decimal_to_hexadecimal(17))
print(decimal_to_hexadecimal(31))


# 내장 함수가 있음
print(bin(5))
print(hex(31))
```

```python
# 2진수를 10진수로 변환
def binary_to_decimal(binary_str):
    decimal_number = 0
    pow = 0

    for digit in reversed(binary_str):
        if digit == "1":
            decimal_number += 2 ** pow
        pow += 1

    return decimal_number


binary_to_decimal("11101")
```

## 16진수와 2진수 변환
16진수를 2진수로 변환할 때, 딕셔너리로 구성하여 풀어도 됨



```python
print(1<<1, bin(1 << 1))

arr = [1, 2, 3, 4, 5, 6, 7]

for i in range(1 << len(arr)):
    subset = []
    total = 0

    for idx in range(len(arr)):
        if i & (1<<idx):
            subset.append(arr[idx])
            total += arr[idx]

    if total == 10:
        print(f'합이 10인 부분집합 : {subset}')
```

```python
# 비트 연산은 어떻게 활용할까?

WALK = 1 << 0
ATTACK = 1 << 1
JUMP = 1 << 2

character_state = 0

def set_state(state, flag):
    return state | flag # 현재 상태는 걷는 상태로 바꿈

def unset_state(state, flag):
    return state & ~flag # 현재 상태는 걷지 않는 상태로 바꿈


set_state(character_state, WALK)
unset_state(character_state, WALK)

```

소수를 2로 곱하여 정수 부분만 사용함.

##### ✔️ 소수점 출력 방법

{t2:.2f} : t2 값을 소수점 셋째자리에서 반올림하여 표현

1. 파이썬에서의 실수 표현 범위
   - 파이썬에서는 64비트 부동소수점으로 실수를 표현함
   - 15 ~ 17자리 정밀도와 +-1.8e308 범위를 갖습니다.
   - 최대로 표현할 수 잇는 값은 약 1.8 * 10^308이고 이 이상은 inf로 표현
   - 최소로 표현할 수 있는(0에 가장 가까운) 값은 약 5.0 * 10^-324이며, 이 이하는 0으로 표현
    ❗파이썬은 동적메모리 활용이 자유롭기 때문에 실수 표현이 자유로움

2. 컴퓨터는 실수를 내부적으로 근사적으로 관리
   - 실수는 정확한 값이 아니라 근사 값으로 저장되는데 이때 생기는 작은 오차가 계산 과정에서 다른 결과를 가져옴
   - 'print(0.1+0.1+0.1 == 0.3)'을 출력하면, 'False'가 나옴
   - 실수는 근사 값이기 때문에 정확히 0.3이 나오지 않으므로 False가 나옴
  

##### ✔️ 실수의 표현
1. 컴퓨터는 실수를 표현하기 위해 부동소수점 표기법을 사용함
    - 이 표기법은 IEEE 754라는 컴퓨터에서 부동소수점을 표기하는 국제표준임
2. 부동 소수점 표기 방법은 소수점의 위치를 고정시켜 표현하는 방식

##### ✔️ 실수를 저장하기 위한 형식
1. 32bit 구조
2. 부호 1비트 : 0이면 양수, 1이면 음수
3. 지수부 : 부동소수점의 크기 + bias 값
4. 가수부 : 실질적 수

```python
T = int(input())

for time in range(1, T+1):

    def binary_to_decimal(binary_number):

        a = str(binary_number).split('.')
        first = a[1]
        result_value = ''
        while first != 0:
            binary_number = binary_number * 2
            result = str(binary_number).split('.')
            binary_number = binary_number - int(result[0])
            result_value += result[0]
            first = int(result[1])

        return result_value

    number = float(input())

    result_2 = binary_to_decimal(number)
    if len(result_2) > 12:
        print(f'#{time} overflow')
    else:
        print(f'#{time} {result_2}')
```
