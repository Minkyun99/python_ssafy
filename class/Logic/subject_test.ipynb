## ⭐ 과목평가 정리
구분	| 부분집합	| 순열	| 조합
|----|----------|------------|-------------|
원리	| 포함/미포함 결정	| 순서를 고려한 나열	| 순서를 고려하지 않은 선택
순서	| 무관	| 중요	| 무관
예시	| {1, 2}의 부분집합: {}, {1}, {2}, {1, 2}	| {1, 2}의 2개 순열: (1, 2), (2, 1) | 	{1, 2}의 2개 조합: {1, 2}
결과	| 집합	| 튜플(Tuple) |	튜플(Tuple)
코드 구현	| 백트래킹, 비트마스킹	| 백트래킹, 라이브러리	| 백트래킹, 라이브러리


```python
# 순열 예시 코드
# 방문한 곳을 저장하여 순서대로 나열하는 경우의 수

def generate_permutations(arr, result, visited):

    if len(result) == len(arr):
        print(result)
        return


    for i in range(len(arr)):

        if not visited[i]:
            visited[i] = 1
            result.append(arr[i])
            generate_permutations(arr, result, visited)
            result.pop()
            visited[i] = 0


my_list = [1, 2, 3]

result = []
visited = [False] * len(my_list)

generate_permutations(my_list, result, visited)
```

```python
# 조합 예시 코드
# 순서에 상관 없이, 중복 없이 나열하는 경우의 수

def generate_combinations(items, k, start_index, current_combination, result_list):

    if len(current_combination) == k:
        result_list.append(current_combination[:]) 
        return

    for i in range(start_index, len(items)):
        # 1. Choose: 현재 원소를 조합에 포함합니다.
        current_combination.append(items[i])
        
        # 2. Recurse: 다음 원소를 선택하기 위해 재귀 호출을 합니다.
        # 다음 탐색은 현재 원소의 다음 인덱스(i + 1)부터 시작합니다.
        # 이는 순서가 바뀌는 중복 조합을 방지합니다.
        generate_combinations(items, k, i + 1, current_combination, result_list)
        
        # 3. Backtrack: 다음 경우의 수를 탐색하기 위해 현재 원소를 제거합니다.
        current_combination.pop()

# --- 예시 사용 ---
my_items = [1, 2, 3, 4]
k = 2  # 4개의 원소 중 2개를 선택

all_combinations = []
generate_combinations(my_items, k, 0, [], all_combinations)

print(f"{my_items}에서 {k}개짜리 조합:")
for combo in all_combinations:
    print(combo)
```

구분	|DFS (깊이 우선 탐색)	|BFS (너비 우선 탐색)
|--|--|--|
|탐색 순서|	한 경로를 끝까지 탐색	|가까운 노드부터 탐색
구현 방식	|재귀, 스택(Stack)	|큐(Queue)
메모리	|적게 사용 (현재 경로만 기억)|	많이 사용 (모든 인접 노드 기억)
활용	|모든 노드 탐색, 사이클 찾기	|최단 경로, 최단 시간
성능	|최단 경로에 비효율적	|최단 경로에 효율적

```python
# 그래프를 딕셔너리로 표현합니다.
# 'A' 노드는 'B', 'C', 'E' 노드와 연결되어 있습니다.
graph = {
    'A': ['B', 'C', 'E'],
    'B': ['A', 'D'],
    'C': ['A', 'G'],
    'D': ['B'],
    'E': ['A', 'F'],
    'F': ['E'],
    'G': ['C']
}

# 방문한 노드를 기록하는 집합(set)입니다.
visited = set()

def dfs(node):
    """
    재귀 함수를 사용하여 깊이 우선 탐색을 수행합니다.

    Args:
        node (str): 현재 방문 중인 노드.
    """
    # 현재 노드를 방문했음을 기록합니다.
    visited.add(node)
    print(node, end=' ') # 노드 방문 순서 출력

    # 현재 노드와 연결된 모든 이웃 노드를 탐색합니다.
    for neighbor in graph[node]:
        # 아직 방문하지 않은 이웃 노드라면 재귀 호출을 통해 탐색을 계속합니다.
        if neighbor not in visited:
            dfs(neighbor)

# 'A' 노드에서 DFS 탐색을 시작합니다.
print("재귀적 DFS 탐색 순서:")
dfs('A')

```

```python
from collections import deque

# 그래프를 딕셔너리로 표현합니다.
# 'A' 노드는 'B', 'C'와 연결되어 있습니다.
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}

def bfs(graph, start_node):
    """
    큐(Queue)를 사용하여 너비 우선 탐색을 수행합니다.

    Args:
        graph (dict): 탐색할 그래프.
        start_node (str): 탐색을 시작할 노드.
    """
    # 방문한 노드를 기록하는 집합(set)입니다.
    visited = set()
    # 방문할 노드를 담는 큐를 생성하고 시작 노드를 추가합니다.
    queue = deque([start_node])

    # 큐가 비어 있지 않은 동안 반복합니다.
    while queue:
        # 큐의 맨 앞에서 노드를 꺼냅니다. (FIFO: 선입선출)
        node = queue.popleft()

        # 아직 방문하지 않은 노드라면
        if node not in visited:
            # 방문했음을 기록하고 출력합니다.
            visited.add(node)
            print(node, end=' ')

            # 현재 노드의 인접 노드들을 큐에 추가합니다.
            # BFS는 너비 우선 탐색이므로, 인접 노드들을 모두 큐에 넣고
            # 다음 턴에서 처리합니다.
            for neighbor in graph[node]:
                if neighbor not in visited:
                    queue.append(neighbor)

    # 'A' 노드에서 BFS 탐색을 시작합니다.
print("BFS 탐색 순서:")
bfs(graph, 'A')
```
