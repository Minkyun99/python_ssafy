## ⭐ 과목평가 정리
구분	| 부분집합	| 순열	| 조합
|----|----------|------------|-------------|
원리	| 포함/미포함 결정	| 순서를 고려한 나열	| 순서를 고려하지 않은 선택
순서	| 무관	| 중요	| 무관
예시	| {1, 2}의 부분집합: {}, {1}, {2}, {1, 2}	| {1, 2}의 2개 순열: (1, 2), (2, 1) | 	{1, 2}의 2개 조합: {1, 2}
결과	| 집합	| 튜플(Tuple) |	튜플(Tuple)
코드 구현	| 백트래킹, 비트마스킹	| 백트래킹, 라이브러리	| 백트래킹, 라이브러리


구분	|DFS (깊이 우선 탐색)	|BFS (너비 우선 탐색)
|--|--|--|
|탐색 순서|	한 경로를 끝까지 탐색	|가까운 노드부터 탐색
구현 방식	|재귀, 스택(Stack)	|큐(Queue)
메모리	|적게 사용 (현재 경로만 기억)|	많이 사용 (모든 인접 노드 기억)
활용	|모든 노드 탐색, 사이클 찾기	|최단 경로, 최단 시간
성능	|최단 경로에 비효율적	|최단 경로에 효율적

```python
# 그래프를 딕셔너리로 표현합니다.
# 'A' 노드는 'B', 'C', 'E' 노드와 연결되어 있습니다.
graph = {
    'A': ['B', 'C', 'E'],
    'B': ['A', 'D'],
    'C': ['A', 'G'],
    'D': ['B'],
    'E': ['A', 'F'],
    'F': ['E'],
    'G': ['C']
}

# 방문한 노드를 기록하는 집합(set)입니다.
visited = set()

def dfs(node):
    """
    재귀 함수를 사용하여 깊이 우선 탐색을 수행합니다.

    Args:
        node (str): 현재 방문 중인 노드.
    """
    # 현재 노드를 방문했음을 기록합니다.
    visited.add(node)
    print(node, end=' ') # 노드 방문 순서 출력

    # 현재 노드와 연결된 모든 이웃 노드를 탐색합니다.
    for neighbor in graph[node]:
        # 아직 방문하지 않은 이웃 노드라면 재귀 호출을 통해 탐색을 계속합니다.
        if neighbor not in visited:
            dfs(neighbor)

# 'A' 노드에서 DFS 탐색을 시작합니다.
print("재귀적 DFS 탐색 순서:")
dfs('A')

```

```python
from collections import deque

# 그래프를 딕셔너리로 표현합니다.
# 'A' 노드는 'B', 'C'와 연결되어 있습니다.
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}

def bfs(graph, start_node):
    """
    큐(Queue)를 사용하여 너비 우선 탐색을 수행합니다.

    Args:
        graph (dict): 탐색할 그래프.
        start_node (str): 탐색을 시작할 노드.
    """
    # 방문한 노드를 기록하는 집합(set)입니다.
    visited = set()
    # 방문할 노드를 담는 큐를 생성하고 시작 노드를 추가합니다.
    queue = deque([start_node])

    # 큐가 비어 있지 않은 동안 반복합니다.
    while queue:
        # 큐의 맨 앞에서 노드를 꺼냅니다. (FIFO: 선입선출)
        node = queue.popleft()

        # 아직 방문하지 않은 노드라면
        if node not in visited:
            # 방문했음을 기록하고 출력합니다.
            visited.add(node)
            print(node, end=' ')

            # 현재 노드의 인접 노드들을 큐에 추가합니다.
            # BFS는 너비 우선 탐색이므로, 인접 노드들을 모두 큐에 넣고
            # 다음 턴에서 처리합니다.
            for neighbor in graph[node]:
                if neighbor not in visited:
                    queue.append(neighbor)

    # 'A' 노드에서 BFS 탐색을 시작합니다.
print("BFS 탐색 순서:")
bfs(graph, 'A')
```
