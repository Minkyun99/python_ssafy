## ⭐ 완전 탐색
모든 가능한 경우를 모두 시도 해본 후, 정답을 찾아내는 알고리즘

- 전체 경우를 모두 돌아보는 경우 -> 시간 복잡도 이슈 발생할 수 있음
-> 시간 복잡도 이슈가 발생하였다면, 가지치기(경우의 수를 줄여보자)

- 규칙 찾기(Greedy) -> 그리디로 해결 할 수 없다면?
- 동적 계획법(DP) : 점화식


```python
path = []
result = 0

def recur(cnt):
    global result, path

    if sum(path) > 10:
        return

    if cnt == 3:
        print(*path)
        result += 1
        return
        
    for num in range(1, 7):
        path.append(num)
        recur(cnt + 1)
        path.pop()


recur(0)
```

```python
path = []
result = 0

def recur(cnt, total):
    global result, path

    if sum(path) > 10:
        return

    if cnt == 3:
        result += 1
        return
    
    for num in range(1, 7):
        recur(cnt + 1, total + num)


recur(0, 0)
print(result)
```
