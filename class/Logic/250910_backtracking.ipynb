##### ✔️ 병합 정렬
⭕ 여러 개의 정렬된 자료의 집합을 병합하여 한 개의 정렬된 집합으로 만드는 방식  
![image.png](./img/logic34.png)  
- 분할/병합 : logn 번 실행  
- 병합 : N번 실행  
=> 시간 복잡도 : Nlogn  

```python
def merge(left, right):
    result = [0] * (len(left) + len(right))
    l = r = 0 # 인덱스

    while l < len(left) and r < len(right):
        if left[l] < right[r]:
            result[l+r] = left[l]
            l += 1
        else:
            result[l + r] = right[r]
            r += 1

    while l < len(left):
        result[l + r] = left[l]
        l += 1

    while r < len(right):
        result[l + r] = right[r]
        r += 1
    
    return result

def merge_sort(li):
    if len(li) == 1:
        return li

    mid = len(li) // 2
    left = li[:mid]
    right = li[mid:]

    # 절반씩 분할
    left_list = merge_sort(left)
    right_list = merge_sort(right)

    merge_list = merge(left_list, right_list)
    return merge_list

arr = [69, 10, 30, 2, 16, 8, 31, 22]

print(merge_sort(arr))
```


##### ✔️ 퀵 정렬
⭕ 기준값을 중심으로 주어진 배열을 두 개로 분할하고, 각각을 정렬하여 전체 배열을 정렬하는 방식  
||병합 정렬|퀵 정렬
|---|---|---|
분할기준|단순히 배열을 반으로 나눔|기준 아이템(pivot item)을 중심으로 기준보다 작은 것을 왼편, 큰 것을 오른편에 위치 시킴
병합처리|정렬된 부분을 다시 병합하는 과정이 필요함|별도의 병합 과정 불필요

- 평균 시간복잡도O(nlogn)
- Partitioning 이라는 과정을 반복하면서, 빠른 속도로 정렬이 되는 알고리즘

1. Partitioning (호어 파티션, 로무토 파티션)
- 작업 영역을 정한다.
- 작업 영역 중 가장 왼쪽에 있는 수를 Pivot이라고 함
- Pivot을 기준으로 왼쪽에는 Pivot 보다 작은 수를 배치, 오른쪽에는 Pivot 보다 큰 수를 배치

2. 호어 파티션
- Pivot을 가장 왼쪽 요소로 선택함
- i와 j를 각각 Pivot을 제외한 가장 왼쪽, 가장 오른쪽 인덱스를 가리킴
- i와 j를 움직이며 Pivot 보다 큰 값과 작은 값을 찾음
- i와 j의 위치를 변경함
- i와 j가 교차하는 순간 Pivot을 해당 위치에 넣어줌  
![image.png](./img/logic35.png)  
![image.png](./img/logic36.png)  
![image.png](./img/logic37.png)  
![image.png](./img/logic38.png)  

```python
def hoare_partition1(arr, left, right):
    """
    Hoare 파티션 스키마를 사용하여 리스트를 분할합니다.
    Args:
        arr (list): 정렬할 리스트.
        left (int): 현재 부분 리스트의 시작 인덱스.
        right (int): 현재 부분 리스트의 끝 인덱스.
    Returns:
        int: 피벗의 최종 위치(j).
    """
    # 피벗을 리스트의 가장 왼쪽 요소로 설정
    pivot = arr[left]
    
    # i는 왼쪽에서 오른쪽으로, j는 오른쪽에서 왼쪽으로 이동하는 포인터
    i = left
    j = right
    
    while i <= j:
        # i는 피벗보다 큰 값을 찾을 때까지 오른쪽으로 이동
        while i <= j and arr[i] <= pivot:
            i += 1
        
        # j는 피벗보다 작은 값을 찾을 때까지 왼쪽으로 이동
        while i <= j and arr[j] >= pivot:
            j -= 1

        # i와 j가 교차하지 않았다면, arr[i]와 arr[j]를 교환
        if i < j:
            arr[i], arr[j] = arr[j], arr[i]

    # 반복문이 끝나면 j는 피벗의 올바른 위치가 됩니다.
    # 피벗(arr[left])과 arr[j]를 교환하여 제자리에 놓습니다.
    # 이 과정이 Hoare 파티션의 핵심입니다.
    arr[left], arr[j] = arr[j], arr[left]
    
    # 피벗의 최종 위치인 j를 반환
    return j

def quick_sort(arr, left, right):
    """
    Hoare 파티션 스키마를 사용하는 퀵 정렬 함수.
    """
    if left < right:
        # 파티션 함수를 호출하여 피벗의 위치를 찾습니다.
        pivot_index = hoare_partition1(arr, left, right)
        
        # 피벗을 기준으로 왼쪽과 오른쪽 부분 리스트를 재귀적으로 정렬
        quick_sort(arr, left, pivot_index - 1)
        quick_sort(arr, pivot_index + 1, right)
        

# 테스트 데이터
arr = [69, 10, 30, 2, 16, 8, 31, 22]

# 퀵 정렬을 호출하여 전체 리스트를 정렬
quick_sort(arr, 0, len(arr) - 1)

# 정렬된 리스트 출력
print("정렬된 리스트:", arr)
```

##### ✔️ 이진 검색
⭕ 자료의 가운데에 있는 항목의 키 값과 비교하여 다음 검색의 위치를 결정하고 검색을 계속 진행하는 방법

1. 이진 검색의 과정
- 자료의 중앙에 있는 원소를 고름
- 중앙 원소의 값과 찾고자 하는 목표 값을 비교함
- 목표 값이 중앙 원소의 값보다 작으면 자료의 왼쪽 반에 대해서 새로 검색을 수행, 크다면 자료의 오른쪽 반에 대해서 새로 검색 수행함
- 찾고자 하는 값을 찾을 때까지 1~3 과정을 반복함

```python
def binary_search_while(target):
    left = 0    # 검색 시작점
    right = len(arr)-1  # 검색 끝점
    cnt = 0 # 몇 번 만에 검색했는가?

    while left <= right:    # 교차되면 못찾은 것
        mid = (left + right) // 2
        cnt += 1

        if arr[mid] == target:
            return mid, cnt  # mid 위치에 존재하면 return
        
        # target 보다 정답이 왼쪽에 있는 경우
        if target < arr[mid]:
            right = mid -1
        # target 보다 정답이 오른쪽에 있는 경우
        else:
            left = mid + 1

arr = [4, 2, 9, 7, 11, 23, 19]
arr.sort()

print(binary_search_while(9))
```

##### ✍️ 분할정복_병합정렬

알고리즘 교수님은 학생들에게 병합 정렬을 이용해 오름차순으로 정렬하는 과제를 내려고 한다.  

정렬 된 결과만으로는 실제로 병합 정렬을 적용했는지 알 수 없기 때문에 다음과 같은 제약을 주었다.  

N개의 정렬 대상을 가진 리스트 L을 분할할 때 L[0:N//2], L[N//2:N]으로 분할 한다.  

병합 과정에서 다음처럼 왼쪽 마지막 원소가 오른쪽 마지막 원소보다 큰 경우의 수를 출력한다.  

![image.png](./img/logic39.png)  

정렬이 끝난 리스트 L에서 L[N//2] 원소를 출력한다.  

알고리즘 교수님의 조건에 따라 병합 정렬을 수행하는 프로그램을 만드시오.  
  
[입력]  

첫 줄에 테스트케이스의 수 T가 주어진다. 1<=T<=50  

다음 줄부터 테스트 케이스의 별로 정수의 개수 N이 주어지고, 다음 줄에 N개의 정수 ai가 주어진다.  

5<=N<=1,000,000, 0 <= ai <= 1,000,000  

[출력]  

각 줄마다 "#T" (T는 테스트 케이스 번호)를 출력한 뒤,  N//2 번째 원소와 오른쪽 원소가 먼저 복사되는 경우의 수를 출력한다.  

```python
def merge(left, right):
    global cnt
    result = [0] * (len(left) + len(right))
    l = r = 0

    while l < len(left) and r < len(right):
        if left[l] < right[r]:
            result[l+r] = left[l]
            l += 1
        else:
            result[l+r] = right[r]
            r += 1


    while l < len(left):
        result[l+r] = left[l]
        l += 1

    while r < len(right):
        result[l+r] = right[r]
        r += 1

    if left[-1] > right[-1]:
        cnt += 1

    return result


def devide(arr):
    if len(arr) == 1:
        return arr
    
    mid = len(arr) // 2
    left = arr[:mid]
    right = arr[mid:]
    

    left_list = devide(left)
    right_list = devide(right)

    merge_list = merge(left_list, right_list)
    return merge_list


T = int(input())

for time in range(1, T+1):
    cnt = 0
    N = int(input())
    arr = list(map(int, input().split()))
    a = devide(arr)
    print(f'#{time} {a[len(a)//2]} {cnt}')

```

##### ✍️ 퀵 정렬

퀵 정렬을 구현해 N개의 정수를 정렬해 리스트 A에 넣고, A[N//2]에 저장된 값을 출력하는 프로그램을 만드시오.  
  
[입력]  

첫 줄에 테스트케이스의 수 T가 주어진다. 1<=T<=50  

다음 줄부터 테스트 케이스의 별로 정수의 개수 N이 주어지고, 다음 줄에 N개의 정수 ai가 주어진다.  

5<=N<=1,000,000, 0 <= ai <= 1,000,000  

[출력]  

각 줄마다 "#T" (T는 테스트 케이스 번호)를 출력한 뒤, , N/2번 원소를 출력한다.  



```python
def hoare_partition1(arr, left, right):
    """
    Hoare 파티션 스키마를 사용하여 리스트를 분할합니다.
    Args:
        arr (list): 정렬할 리스트.
        left (int): 현재 부분 리스트의 시작 인덱스.
        right (int): 현재 부분 리스트의 끝 인덱스.
    Returns:
        int: 피벗의 최종 위치(j).
    """
    # 피벗을 리스트의 가장 왼쪽 요소로 설정
    pivot = arr[left]
    
    # i는 왼쪽에서 오른쪽으로, j는 오른쪽에서 왼쪽으로 이동하는 포인터
    i = left
    j = right
    
    while i <= j:
        # i는 피벗보다 큰 값을 찾을 때까지 오른쪽으로 이동
        while i <= j and arr[i] <= pivot:
            i += 1
        
        # j는 피벗보다 작은 값을 찾을 때까지 왼쪽으로 이동
        while i <= j and arr[j] >= pivot:
            j -= 1

        # i와 j가 교차하지 않았다면, arr[i]와 arr[j]를 교환
        if i < j:
            arr[i], arr[j] = arr[j], arr[i]

    # 반복문이 끝나면 j는 피벗의 올바른 위치가 됩니다.
    # 피벗(arr[left])과 arr[j]를 교환하여 제자리에 놓습니다.
    # 이 과정이 Hoare 파티션의 핵심입니다.
    arr[left], arr[j] = arr[j], arr[left]
    
    # 피벗의 최종 위치인 j를 반환
    return j

def quick_sort(arr, left, right):
    """
    Hoare 파티션 스키마를 사용하는 퀵 정렬 함수.
    """
    if left < right:
        # 파티션 함수를 호출하여 피벗의 위치를 찾습니다.
        pivot_index = hoare_partition1(arr, left, right)
        
        # 피벗을 기준으로 왼쪽과 오른쪽 부분 리스트를 재귀적으로 정렬
        quick_sort(arr, left, pivot_index - 1)
        quick_sort(arr, pivot_index + 1, right)

T = int(input())

for time in range(1, T+1):
    N = int(input())
    # 테스트 데이터
    arr = list(map(int, input().split()))

    # 퀵 정렬을 호출하여 전체 리스트를 정렬
    quick_sort(arr, 0, len(arr) - 1)

    # 정렬된 리스트 출력
    print(f'#{time} {arr[len(arr)//2]}')
```

##### ✍️ 분할정복_이진탐색

서로 다른 정수 N개가 주어지면 정렬한 상태로 리스트 A에 저장한다. 그런 다음 리스트 B에 저장된 M개의 정수에 대해 A에 들어있는 수인지 이진 탐색을 통해 확인하려고 한다.  

전체 탐색 구간의 시작과 끝 인덱스를 l과 r이라고 하면, 중심 원소의 인덱스 m=(l+r)//2 이고, 이진 탐색의 왼쪽 구간은 l부터 m-1, 오른쪽 구간은 m+1부터 r이 된다.  

이때 B에 속한 어떤 수가 A에 들어있으면서, 동시에 탐색 과정에서 양쪽구간을 번갈아 선택하게 되는 숫자의 개수를 알아보려고 한다.  

다음은 10개의 정수가 저장된 리스트 A에서 이진 탐색으로 6을 찾는 예이다.  
   
 ![image.png](./img/logic40.png)  

6은 탐색 과정에서 양쪽을 번갈아 가며 선택하게 된다.   

예를 들어 10을 찾는 경우 오른쪽-오른쪽 구간을 선택하므로 조건에 맞지 않는다.  

5를 찾는 경우 m에 위치하므로 조건에 맞는다.  

이때 m에 찾는 원소가 있는 경우 방향을 따지지 않는다.  
M개의 정수 중 조건을 만족하는 정수의 개수를 알아내는 프로그램을 만드시오.  
  

[입력]  

첫 줄에 테스트케이스의 수 T가 주어진다. 1<=T<=50  

다음 줄부터 테스트 케이스의 별로 A와 B에 속한 정수의 개수 N, M이 주어지고,   
두 줄에 걸쳐 N개와 M개의 백만 이하의 양의 정수가 주어진다.  

1<=N, M<=500,000  

[출력]  

각 줄마다 "#T" (T는 테스트 케이스 번호)를 출력한 뒤, 답을 출력한다.  

```python
def binary(arr, target):
    
    left = 0
    right = len(arr) - 1
    is_True = True
    
    while left <= right:
        mid = (left + right) // 2
        
        if arr[mid] == target:
            return mid, is_True
        
        if target < arr[mid]:
            right = mid - 1

            Next_mid = (left + right) // 2
            if target < arr[Next_mid]:
                is_True = False

        else:
            left = mid + 1

            Next_mid = (left + right) // 2
            if target > arr[Next_mid]:
                is_True = False

T = int(input())

for time in range(1, T+1):

    N, M = map(int, input().split())
    arr = list(map(int, input().split()))
    arr = sorted(arr)
    li = list(map(int, input().split()))
    count = 0


    for i in li:
        A = binary(arr, i)
        if A and A[1] == True:
            count += 1

    print(f'#{time} {count}')





```
